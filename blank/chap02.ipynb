Neste exemplo, a expressão é um inteiro.

## Nomes de variáveis

Nomes de variáveis podem ser tão longos quanto você quiser. Podem conter letras e números, mas não podem começar com um número.
É válido usar letras maiúsculas, mas o convencional é usar só minúsculas em
nomes de variáveis.

A única pontuação que pode aparecer num nome de variável é o caractere de sublinhado, `_`. Ele costuma ser usado em nomes com várias palavras, como `your_name` ou `airspeed_of_unladen_swallow`.

Se você der a uma variável um nome inválido, recebe um erro de sintaxe.
O nome `million!` é inválido porque contém pontuação.

`76trombones` é inválido porque começa com um número.

`class` também é inválido, mas o motivo pode não ser óbvio.

Acontece que `class` é uma **palavra-chave**, uma palavra especial usada para especificar a estrutura de um programa.
Palavras-chave não podem ser usadas como nomes de variáveis.

Aqui está a lista completa das palavras-chave de Python:

```
False      await      else       import     pass
None       break      except     in         raise
True       class      finally    is         return
and        continue   for        lambda     try
as         def        from       nonlocal   while
assert     del        global     not        with
async      elif       if         or         yield
```

Você não precisa memorizar essa lista. Na maior parte dos ambientes de desenvolvimento,
as palavras-chave aparecem numa cor diferente; se você tentar usar uma como
nome de variável, vai perceber.

## Expressões e instruções

Até agora, vimos alguns tipos de expressões.
Uma expressão pode ser um valor só, como um inteiro, um número de ponto flutuante ou uma string.
Também pode ser uma coleção de valores e operadores.
E pode incluir nomes de variáveis e chamadas de função.
Aqui está uma expressão que inclui vários desses elementos.

Também vimos alguns tipos de instruções.
Uma **instrução** é uma unidade de código que tem um efeito, mas não tem valor.
Por exemplo, uma instrução de atribuição cria uma variável e lhe dá um valor, mas a instrução em si não tem valor.

Da mesma forma, uma instrução de importação tem um efeito: ela importa um módulo para que possamos usar as variáveis e funções que ele contém, mas não tem efeito visível.

Calcular o valor de uma expressão se chama **avaliação**.
Executar uma instrução se chama **execução**.

## A função print

Quando você avalia uma expressão, o resultado é exibido.

Mas se você avalia mais de uma expressão, só o valor da última é exibido.

Para exibir mais de um valor, você pode usar a função `print`.

Ela também funciona com números de ponto flutuante e strings.

Você também pode usar uma sequência de expressões separadas por vírgulas.

Observe que a função `print` coloca um espaço entre os valores.

## Argumentos

Quando você chama uma função, a expressão entre parênteses se chama **argumento**.
Normalmente eu explicaria o porquê, mas neste caso o sentido técnico do termo quase não tem relação com o sentido comum da palavra, então nem vou tentar.

Algumas das funções que vimos até agora recebem só um argumento, como `int`.

Algumas recebem dois, como `math.pow`.

Algumas podem receber argumentos adicionais, que são opcionais.
Por exemplo, `int` pode receber um segundo argumento que especifica a base do número.

A sequência de dígitos `101` na base 2 representa o número 5 na base 10.

`round` também recebe um segundo argumento opcional, que é o número de casas decimais para as quais arredondar.

Algumas funções podem receber qualquer quantidade de argumentos, como `print`.

Se você chama uma função e fornece argumentos demais, isso é um `TypeError`.

Se você fornece argumentos de menos, isso também é um `TypeError`.

E se você fornece um argumento com um tipo que a função não consegue tratar, isso também é um `TypeError`.

Esse tipo de verificação pode irritar no começo, mas ajuda a detectar e corrigir erros.

## Depuração

Três tipos de erro podem ocorrer num programa: erros de sintaxe, erros de tempo de execução e erros semânticos.
É útil distingui-los para rastreá-los mais depressa.

* **Erro de sintaxe**: "sintaxe" se refere à estrutura de um programa e às regras sobre essa estrutura. Se houver um erro de sintaxe em qualquer lugar do programa, o Python não o executa. Ele exibe uma mensagem de erro imediatamente.

* **Erro de tempo de execução**: se não houver erros de sintaxe no programa, ele pode começar a executar. Mas se algo der errado, o Python exibe uma mensagem de erro e para. Esse tipo de erro se chama erro de tempo de execução. Também se chama **exceção** porque indica que algo excepcional aconteceu.

* **Erro semântico**: o terceiro tipo de erro é "semântico", o que significa relacionado ao sentido. Se houver um erro semântico no programa, ele executa sem gerar mensagens de erro, mas não faz o que você pretendia. Identificar erros semânticos pode ser complicado, porque exige trabalhar de trás para frente, olhando a saída do programa e tentando descobrir o que ele está fazendo.

Como vimos, um nome de variável inválido é um erro de sintaxe.

Se você usa um operador com um tipo que ele não admite, isso é um erro de tempo de execução.

## Glossário

**variável:**
Um nome que se refere a um valor.

**instrução de atribuição:**
Uma instrução que atribui um valor a uma variável.

**diagrama de estado:**
Uma representação gráfica de um conjunto de variáveis e dos valores a que elas se referem.

**palavra-chave:**
Uma palavra especial usada para especificar a estrutura de um programa.

**instrução import:**
Uma instrução que lê um arquivo de módulo para que possamos usar as variáveis e funções que ele contém.

**módulo:**
Um arquivo que contém código Python, inclusive definições de função e às vezes outras instruções.

**operador ponto:**
O operador `.`, usado para acessar uma função em outro módulo especificando o nome do módulo, seguido de um ponto e do nome da função.

**avaliar:**
Realizar as operações de uma expressão para calcular um valor.

**instrução:**
Uma ou mais linhas de código que representam um comando ou uma ação.

**executar:**
Rodar uma instrução e fazer o que ela diz.

**argumento:**
Um valor fornecido a uma função quando a função é chamada.

**comentário:**
Texto incluído num programa que fornece informação sobre o programa, mas não tem efeito na execução.

**erro de tempo de execução:**
Um erro que faz o programa exibir uma mensagem de erro e encerrar.

**exceção:**
Um erro detectado enquanto o programa está em execução.

**erro semântico:**
Um erro que faz o programa fazer a coisa errada, mas sem exibir uma mensagem de erro.
