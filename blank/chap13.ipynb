# Arquivos e bancos de dados

A maior parte dos programas que vimos até agora é **efêmera** no sentido de que roda por um tempo curto e produz uma saída, mas quando termina, os dados desaparecem.
Cada vez que você executa um programa efêmero, ele começa do zero.

Outros programas são **persistentes**: rodam por muito tempo (ou o tempo todo); guardam pelo menos parte dos dados em armazenamento de longo prazo; e se forem encerrados e reiniciados, retomam de onde pararam.

Um jeito simples de os programas manterem os dados é lendo e gravando arquivos de texto.
Uma alternativa mais versátil é guardar os dados em um banco de dados.
Bancos de dados são arquivos especializados que podem ser lidos e gravados com mais eficiência do que arquivos de texto, e oferecem capacidades adicionais.

Neste capítulo, vamos escrever programas que leem e gravam arquivos de texto e bancos de dados e, como exercício, você vai escrever um programa que busca duplicatas em uma coleção de fotos.
Mas antes de trabalhar com um arquivo, você precisa encontrá-lo, então vamos começar com nomes de arquivo, caminhos e diretórios.

## Nomes de arquivo e caminhos

Os arquivos são organizados em **diretórios**, também chamados de "pastas".
Todo programa em execução tem um **diretório de trabalho atual**, que é o diretório padrão para a maior parte das operações.
Por exemplo, quando você abre um arquivo, o Python o procura no diretório de trabalho atual.

O módulo `os` oferece funções para trabalhar com arquivos e diretórios ("os" vem de "operating system"). 
Ele oferece uma função chamada `getcwd` que obtém o nome do diretório de trabalho atual.

O resultado neste exemplo é o diretório home de um usuário chamado `dinsdale`.
Uma string como `'/home/dinsdale'` que identifica um arquivo ou diretório se chama **caminho**.

Um nome de arquivo simples como `'memo.txt'` também é considerado um caminho, mas é um **caminho relativo** porque especifica um nome de arquivo em relação ao diretório atual.
Neste exemplo, o diretório atual é `/home/dinsdale`, então `'memo.txt'` equivale ao caminho completo `'/home/dinsdale/memo.txt'`.

Um caminho que começa com `/` não depende do diretório atual: chama-se **caminho absoluto**. 
Para encontrar o caminho absoluto de um arquivo, você pode usar `abspath`.

O módulo `os` oferece outras funções para trabalhar com nomes de arquivo e caminhos.
`listdir` devolve uma lista do conteúdo do diretório dado, incluindo arquivos e outros diretórios.
Aqui está um exemplo que lista o conteúdo de um diretório chamado `photos`.

Este diretório contém um arquivo de texto chamado `notes.txt` e três diretórios.
Os diretórios contêm arquivos de imagem no formato JPEG.

Para verificar se um arquivo ou diretório existe, podemos usar `os.path.exists`.

Para verificar se um caminho se refere a um arquivo ou a um diretório, podemos usar `isdir`, que devolve `True` se o caminho se refere a um diretório.

E `isfile`, que devolve `True` se o caminho se refere a um arquivo.

Um desafio de trabalhar com caminhos é que eles têm aparência diferente em sistemas operacionais diferentes.
No macOS e em sistemas UNIX como o Linux, os nomes de diretório e de arquivo em um caminho são separados por uma barra, `/`.
O Windows usa uma barra invertida, `\`.
Assim, se você rodar estes exemplos no Windows, verá barras invertidas nos caminhos e terá de substituir as barras dos exemplos.

Ou, para escrever código que funcione nos dois sistemas, você pode usar `os.path.join`, que junta nomes de diretório e de arquivo em um caminho usando barra ou barra invertida, conforme o sistema operacional.

Mais adiante neste capítulo, vamos usar essas funções para percorrer um conjunto de diretórios e encontrar todos os arquivos de imagem.

## Shelve

Até agora temos lido e gravado arquivos de texto. Agora vamos considerar bancos de dados.
Um **banco de dados** é um arquivo organizado para guardar dados.
Alguns bancos de dados são organizados como uma tabela com linhas e colunas de informação.
Outros são organizados como um dicionário que mapeia de chaves para valores; às vezes são chamados de **armazenamentos chave-valor**.

O módulo `shelve` oferece funções para criar e atualizar um armazenamento chave-valor chamado "shelf".
Como exemplo, vamos criar um shelf para conter legendas das figuras no diretório `photos`.
Vamos usar o dicionário `config` para obter o nome do diretório em que devemos colocar o shelf.

Podemos usar `os.makedirs` para criar este diretório, se ele ainda não existir.

E `os.path.join` para montar um caminho que inclui o nome do diretório e o nome do arquivo do shelf, `captions`.

Agora podemos usar `shelve.open` para abrir o arquivo do shelf.
O argumento `c` indica que o arquivo deve ser criado se necessário.

O valor de retorno é oficialmente um objeto `DbfilenameShelf`, mais informalmente chamado de objeto shelf.

O objeto shelf se comporta em muitos aspectos como um dicionário.
Por exemplo, podemos usar o operador de colchetes para adicionar um item, que é um mapeamento de uma chave para um valor.

Neste exemplo, a chave é o caminho de um arquivo de imagem e o valor é uma string que descreve a imagem.

Também usamos o operador de colchetes para procurar uma chave e obter o valor correspondente.

Se você fizer outra atribuição a uma chave existente, `shelve` substitui o valor antigo.

Alguns métodos de dicionário, como `keys`, `values` e `items`, também funcionam com objetos shelf.

Podemos usar o operador `in` para verificar se uma chave aparece no shelf.

E podemos usar uma instrução `for` para percorrer as chaves.

Como com outros arquivos, você deve fechar o banco de dados quando terminar.

Agora, se listarmos o conteúdo do diretório de dados, vemos dois arquivos.

`captions.dat` contém os dados que acabamos de guardar.
`captions.dir` contém informações sobre a organização do banco de dados que tornam o acesso mais eficiente.
O sufixo `dir` vem de "directory", mas não tem nada a ver com os diretórios com que temos trabalhado, que contêm arquivos.

## Verificando arquivos equivalentes

Agora voltemos ao objetivo deste capítulo: buscar arquivos diferentes que contêm os mesmos dados.
Um jeito de verificar é ler o conteúdo dos dois arquivos e comparar.

Se os arquivos contêm imagens, temos de abri-los com o modo `'rb'`, em que `'r'` significa que queremos ler o conteúdo e `'b'` indica **modo binário**.
No modo binário, o conteúdo não é interpretado como texto: é tratado como uma sequência de bytes.

Esses dois arquivos não são equivalentes.

Se temos só dois arquivos, esta função é uma boa opção.
Mas suponha que temos um grande número de arquivos e queremos saber se algum par deles contém os mesmos dados.
Seria ineficiente comparar cada par de arquivos.

Uma alternativa é usar uma **função de hash**, que recebe o conteúdo de um arquivo e calcula um **digest**, que em geral é um inteiro grande.
Se dois arquivos contêm os mesmos dados, terão o mesmo digest.
Se dois arquivos diferem, terão *quase sempre* digests diferentes.

O módulo `hashlib` oferece várias funções de hash: a que vamos usar se chama `md5`.
Vamos começar usando `hashlib.md5` para criar um objeto `HASH`.

O objeto `HASH` oferece um método `update` que recebe o conteúdo do arquivo como argumento.

Agora podemos usar `hexdigest` para obter o digest como uma string de dígitos hexadecimais que representam um inteiro na base 16.

## Depuração

Quando você está lendo e gravando arquivos, pode esbarrar em problemas com espaço em branco.
Esses erros podem ser difíceis de depurar porque os caracteres de espaço em branco normalmente são invisíveis.
Por exemplo, aqui está uma string que contém espaços, uma tabulação representada pela sequência `\t` e uma quebra de linha representada pela sequência `\n`.
Quando a imprimimos, não vemos os caracteres de espaço em branco.

A função nativa `repr` pode ajudar. Ela recebe qualquer objeto como argumento e devolve uma representação em string do objeto.
Para strings, representa caracteres de espaço em branco com sequências de barra invertida.

Isso pode ser útil para depurar.

Outro problema com que você pode se deparar é que sistemas diferentes usam caracteres diferentes para indicar o fim de uma linha. Alguns sistemas usam uma quebra de linha, representada por `\n`. Outros usam um caractere de retorno, representado por `\r`. 
Alguns usam os dois. Se você mover arquivos entre sistemas diferentes, essas
inconsistências podem causar problemas.

A capitalização de nomes de arquivo é outra questão que você pode encontrar se trabalhar com sistemas operacionais diferentes.
No macOS e no UNIX, os nomes de arquivo podem conter letras minúsculas e maiúsculas, dígitos e a maior parte dos símbolos.
Mas muitos aplicativos do Windows ignoram a diferença entre minúsculas e maiúsculas, e vários símbolos permitidos no macOS e no UNIX não são permitidos no Windows.

## Glossário

**efêmero:**
Um programa efêmero em geral roda por um tempo curto e, quando termina, seus dados se perdem.

**persistente:**
 Um programa persistente roda indefinidamente e guarda pelo menos parte dos dados em armazenamento permanente.

**diretório:**
Uma coleção de arquivos e outros diretórios.

**diretório de trabalho atual:**
O diretório padrão usado por um programa, a menos que outro diretório seja especificado.

**caminho:**
 Uma string que especifica uma sequência de diretórios, muitas vezes levando a um arquivo.

**caminho relativo:**
Um caminho que parte do diretório de trabalho atual, ou de algum outro diretório especificado.

**caminho absoluto:**
Um caminho que não depende do diretório atual.

**f-string:**
Uma string que tem a letra `f` antes das aspas de abertura e contém uma ou mais expressões entre chaves.

**dados de configuração:**
Dados, muitas vezes guardados em um arquivo, que especificam o que um programa deve fazer e como.

**serialização:**
Converter um objeto em uma string.

**desserialização:**
Converter uma string em um objeto.

**banco de dados:**
 Um arquivo cujo conteúdo é organizado para realizar certas operações com eficiência.

**armazenamentos chave-valor:**
Um banco de dados cujo conteúdo é organizado como um dicionário, com chaves que correspondem a valores.

**modo binário:**
Um jeito de abrir um arquivo para que o conteúdo seja interpretado como uma sequência de bytes, e não como uma sequência de caracteres.

**função de hash:**
Uma função que recebe um objeto e calcula um inteiro, às vezes chamado de digest.

**digest:**
O resultado de uma função de hash, sobretudo quando é usado para verificar se dois objetos são iguais.
