## Repetição

Se quisermos exibir mais de um verso, podemos usar uma instrução `for`.
Aqui está um exemplo simples.

A primeira linha é um cabeçalho que termina com dois-pontos.
A segunda linha é o corpo, que precisa estar recuado.

O cabeçalho começa com a palavra-chave `for`, uma variável nova chamada `i` e outra palavra-chave, `in`.
Ele usa a função `range` para criar uma sequência de dois valores, que são `0` e `1`.
Em Python, quando começamos a contar, em geral começamos do `0`.

Quando a instrução `for` executa, ela atribui o primeiro valor de `range` a `i` e depois executa a função `print` no corpo, que exibe `0`.

Quando chega ao fim do corpo, ela volta ao cabeçalho, e por isso esta instrução se chama **laço**.
Na segunda passagem pelo laço, ela atribui o próximo valor de `range` a `i` e o exibe.
Depois, como esse é o último valor de `range`, o laço termina.

Eis como podemos usar um laço `for` para imprimir dois versos da música.

Você pode colocar um laço `for` dentro de uma função.
Por exemplo, `print_n_verses` recebe um parâmetro chamado `n`, que precisa ser um inteiro, e exibe a quantidade dada de versos.

Neste exemplo, não usamos `i` no corpo do laço, mas mesmo assim precisa haver um nome de variável no cabeçalho.

## Variáveis e parâmetros são locais

Quando você cria uma variável dentro de uma função, ela é **local**, o que
significa que só existe dentro da função.
Por exemplo, a função a seguir recebe dois argumentos, concatena-os e imprime o resultado duas vezes.

Aqui está um exemplo que a usa:

Quando `cat_twice` executa, ela cria uma variável local chamada `cat`, que é destruída quando a função termina.

Parâmetros também são locais.

## Diagramas de pilha

Para acompanhar quais variáveis podem ser usadas onde, às vezes é útil desenhar um **diagrama de pilha**.
Como os diagramas de estado, os diagramas de pilha mostram o valor de cada variável, mas também mostram a função a que cada variável pertence.

Cada função é representada por um **quadro**.
Um quadro é uma caixa com o nome de uma função do lado de fora e os parâmetros e variáveis locais da função do lado de dentro.

Aqui está o diagrama de pilha do exemplo anterior.

````python
from diagram import make_frame, Stack

d1 = dict(line1=line1, line2=line2)
frame1 = make_frame(d1, name='__main__', dy=-0.3, loc='left')

d2 = dict(part1=line1, part2=line2, cat=line1+line2)
frame2 = make_frame(d2, name='cat_twice', dy=-0.3, 
                    offsetx=0.03, loc='left')

d3 = dict(string=line1+line2)
frame3 = make_frame(d3, name='print_twice', 
                    offsetx=0.04, offsety=-0.3, loc='left')

d4 = {"?": line1+line2}
frame4 = make_frame(d4, name='print', 
                    offsetx=-0.22, offsety=0, loc='left')

stack = Stack([frame1, frame2, frame3, frame4], dy=-0.8)
````

````python
from diagram import diagram, adjust


width, height, x, y = [3.77, 2.9, 1.1, 2.65]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)
````

Os quadros são dispostos numa pilha que indica qual função chamou
qual, e assim por diante. Lendo de baixo para cima, `print` foi chamada por `print_twice`, que foi chamada por `cat_twice`, que foi chamada por `__main__`, que é um nome especial para o quadro mais alto.
Quando você cria uma variável fora de qualquer função, ela pertence a `__main__`.

No quadro de `print`, o ponto de interrogação indica que não sabemos o nome do parâmetro.

## Por que funções?

Pode ainda não estar claro por que vale o trabalho de dividir um programa em
funções.
Há várias razões:

-   Criar uma função nova dá a oportunidade de nomear um grupo de
    instruções, o que deixa o programa mais fácil de ler e depurar.

-   Funções podem deixar um programa menor ao eliminar código repetitivo.
    Depois, se você fizer uma mudança, só precisa fazê-la num lugar.

-   Dividir um programa longo em funções permite depurar as partes
    uma de cada vez e depois montá-las num todo que funciona.

-   Funções bem projetadas costumam ser úteis em muitos programas. Depois que você
    escreve e depura uma, pode reutilizá-la.

## Glossário

**definição de função:**
Uma instrução que cria uma função.

**cabeçalho:**
 A primeira linha de uma definição de função.

**corpo:**
 A sequência de instruções dentro de uma definição de função.

**objeto de função:**
Um valor criado por uma definição de função.
O nome da função é uma variável que se refere a um objeto de função.

**parâmetro:**
 Um nome usado dentro de uma função para se referir ao valor passado como argumento.

**laço:**
 Uma instrução que executa uma ou mais instruções, muitas vezes de forma repetida.

**variável local:**
Uma variável definida dentro de uma função, e que só pode ser acessada dentro da função.

**diagrama de pilha:**
Uma representação gráfica de uma pilha de funções, suas variáveis e os valores a que elas se referem.

**quadro:**
 Uma caixa num diagrama de pilha que representa uma chamada de função.
 Contém as variáveis locais e os parâmetros da função.

**traceback:**
 Uma lista das funções que estão em execução, impressa quando ocorre uma exceção.
