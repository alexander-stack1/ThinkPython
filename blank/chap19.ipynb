# Considerações finais

Aprender a programar não é fácil, mas, se você chegou até aqui, já deu um bom começo.
Agora tenho algumas sugestões de como continuar aprendendo e aplicar o que aprendeu.

Este livro pretende ser uma introdução geral à programação, então não nos concentramos em aplicações específicas.
Conforme seus interesses, há várias áreas em que você pode aplicar as habilidades novas.

Se você se interessa por ciência de dados, há três livros meus que podem agradar:

* *Think Stats: Exploratory Data Analysis*, O'Reilly Media, 2014.

* *Think Bayes: Bayesian Statistics in Python*, O'Reilly Media, 2021.

* *Think DSP: Digital Signal Processing in Python*, O'Reilly Media, 2016.

Se você se interessa por modelagem física e sistemas complexos, pode gostar de:

* *Modeling and Simulation in Python: An Introduction for Scientists and Engineers*, No Starch Press, 2023.

* *Think Complexity: Complexity Science and Computational Modeling*, O'Reilly Media, 2018.

Eles usam NumPy, SciPy, pandas e outras bibliotecas de Python para ciência de dados e computação científica.

Este livro tenta equilibrar princípios gerais de programação e detalhes de Python.
Por isso, não inclui todos os recursos da linguagem.
Para saber mais sobre Python, e bons conselhos sobre como usá-lo, recomendo *Fluent Python: Clear, Concise, and Effective Programming*, segunda edição, de Luciano Ramalho, O'Reilly Media, 2022.

Depois de uma introdução à programação, o próximo passo comum é aprender estruturas de dados e algoritmos.
Tenho um trabalho em andamento sobre esse tema, chamado *Data Structures and Information Retrieval in Python*.
Uma versão eletrônica gratuita está disponível na Green Tea Press em <https://greenteapress.com>.

À medida que trabalhar em programas mais complexos, você vai encontrar desafios novos.
Pode ser útil reler as seções deste livro sobre depuração.
Em especial, lembre-se dos seis erres da depuração no [capítulo 12](section_debugging_12): ler, rodar, ruminar, explicar ao patinho de borracha, recuar e descansar.

Este livro sugere ferramentas para ajudar na depuração, inclusive as funções `print` e `repr`, a função `structshape` no [capítulo 11](section_debugging_11), e as funções nativas `isinstance`, `hasattr` e `vars` no [capítulo 14](section_debugging_14).

Também sugere ferramentas para testar programas, inclusive a instrução `assert`, o módulo `doctest` e o módulo `unittest`.
Incluir testes nos programas é uma das melhores formas de prevenir e detectar erros, e de economizar tempo na depuração.

Mas o melhor tipo de depuração é aquele que você não precisa fazer.
Se usar um processo de desenvolvimento incremental como o descrito no [capítulo 6](section_incremental), e testar à medida que avança, você vai cometer menos erros e encontrá-los mais depressa quando acontecerem.
Lembre-se também de encapsulamento e generalização no [capítulo 4](section_encapsulation), especialmente úteis quando você desenvolve código em notebooks Jupyter.

Ao longo deste livro, sugeri formas de usar assistentes virtuais para ajudar a aprender, programar e depurar.
Espero que você esteja achando essas ferramentas úteis.

Além de assistentes virtuais como o ChatGPT, você também pode querer usar uma ferramenta como o Copilot, que completa código enquanto você digita.
No começo eu não recomendei essas ferramentas, porque podem ser avassaladoras para quem está começando.
Mas agora você pode querer explorá-las.

Usar ferramentas de IA com eficácia exige alguma experimentação e reflexão para encontrar um fluxo que funcione para você.
Se achar um incômodo copiar código do ChatGPT para o Jupyter, talvez prefira algo como o Copilot.
Mas o trabalho cognitivo de compor um prompt e interpretar a resposta pode ser tão valioso quanto o código que a ferramenta gera, no mesmo espírito da depuração com o patinho de borracha.

Conforme ganhar experiência em programação, você pode querer explorar outros ambientes de desenvolvimento.
Acho que notebooks Jupyter são um bom ponto de partida, mas são relativamente novos e não tão usados quanto os ambientes de desenvolvimento integrado (IDEs) convencionais.
Para Python, os IDEs mais populares incluem PyCharm e Spyder, e o Thonny, que costuma ser recomendado para iniciantes.
Outros IDEs, como Visual Studio Code e Eclipse, também funcionam com outras linguagens de programação.
Ou, como alternativa mais simples, você pode escrever programas em Python em qualquer editor de texto de que gostar.

Na continuação da jornada de programação, você não precisa ir sozinho!
Se mora em uma cidade ou perto de uma, é bem provável que exista um grupo de usuários de Python do qual você possa participar.
Esses grupos em geral são receptivos a iniciantes, então não tenha medo.
Se não houver um grupo perto de você, talvez consiga participar de eventos à distância.
Fique de olho também em conferências regionais de Python.

Uma das melhores formas de melhorar as habilidades de programação é aprender outra linguagem.
Se você se interessa por estatística e ciência de dados, pode querer aprender R.
Mas recomendo em especial aprender uma linguagem funcional como Racket ou Elixir.
A programação funcional exige um tipo diferente de pensamento, o que muda o modo como você pensa sobre programas.

Boa sorte!
