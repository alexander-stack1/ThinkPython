# Listas

Este capítulo apresenta um dos tipos nativos mais úteis de Python, as listas.
Você também vai aprender mais sobre objetos e o que pode acontecer quando várias variáveis se referem ao mesmo objeto.

## Uma lista é uma sequência

Como uma string, uma **lista** é uma sequência de valores. Em uma string, os
valores são caracteres; em uma lista, eles podem ser de qualquer tipo.
Os valores de uma lista se chamam **elementos**.

Há vários jeitos de criar uma lista nova; o mais simples é delimitar os elementos com colchetes (`[` e `]`).
Por exemplo, aqui está uma lista de dois inteiros.

E aqui está uma lista de três strings.

Os elementos de uma lista não precisam ser do mesmo tipo.
A lista a seguir contém uma string, um float, um inteiro e até outra lista.

Uma lista dentro de outra lista é **aninhada**.

Uma lista que não contém nenhum elemento se chama lista vazia; você pode criar
uma com colchetes vazios, `[]`.

A função `len` devolve o comprimento de uma lista.

O comprimento de uma lista vazia é `0`.

## Listas são mutáveis

Para ler um elemento de uma lista, podemos usar o operador de colchetes.
O índice do primeiro elemento é `0`.

Diferente das strings, listas são mutáveis. Quando o operador de colchetes aparece do
lado esquerdo de uma atribuição, ele identifica o elemento da lista
que será atribuído.

O segundo elemento de `numbers`, que era `123`, agora é `17`.

Os índices de lista funcionam do mesmo jeito que os índices de string:

-   Qualquer expressão inteira pode ser usada como índice.

-   Se você tentar ler ou escrever um elemento que não existe, recebe
    um `IndexError`.

-   Se um índice tiver um valor negativo, ele conta de trás para a frente a partir do fim da
    lista.

O operador `in` funciona com listas: ele verifica se um elemento dado aparece em algum lugar da lista.

Embora uma lista possa conter outra lista, a lista aninhada ainda conta como um único elemento: então, na lista a seguir, há só quatro elementos.

E `10` não é considerado um elemento de `t` porque é um elemento de uma lista aninhada, não de `t`.

O operador de fatia funciona em listas do mesmo jeito que funciona em strings.

Se você omitir o primeiro índice, a fatia começa no início.

## Operações com listas

O operador `+` concatena listas.

O operador `*` repete uma lista um número dado de vezes.

Nenhum outro operador matemático funciona com listas, mas a função nativa `sum` soma os elementos.

E `min` e `max` acham o menor e o maior elemento.

## Apelidamento

Se `a` se refere a um objeto e você atribui `b = a`, então as duas variáveis se referem ao mesmo objeto.

Então o diagrama de estado fica assim.

A associação de uma variável a um objeto se chama **referência**.
Neste exemplo, há duas referências ao mesmo objeto.

Um objeto com mais de uma referência tem mais de um nome, então dizemos que o objeto está **apelidado**.
Se o objeto apelidado for mutável, mudanças feitas com um nome afetam o outro.
Neste exemplo, se mudarmos o objeto a que `b` se refere, também estamos mudando o objeto a que `a` se refere.

Então diríamos que `a` "vê" essa mudança.
Embora este comportamento possa ser útil, ele é propenso a erros.
Em geral, é mais seguro evitar o apelidamento quando você trabalha com objetos mutáveis.

Para objetos imutáveis como strings, o apelidamento não é um problema tão grande.

## Glossário

**lista:**
 Um objeto que contém uma sequência de valores.

**elemento:**
 Um dos valores de uma lista ou de outra sequência.

**lista aninhada:**
Uma lista que é um elemento de outra lista.

**delimitador:**
 Um caractere ou string usado para indicar onde uma string deve ser quebrada.

**equivalente:**
 Ter o mesmo valor.

**idêntico:**
 Ser o mesmo objeto (o que implica equivalência).

**referência:**
 A associação entre uma variável e o seu valor.

**apelidado:**
Se há mais de uma variável que se refere a um objeto, o objeto está apelidado.

**atributo:**
 Um dos valores nomeados associados a um objeto.
