# Classes e métodos

Python é uma **linguagem orientada a objetos**: isto é, oferece recursos que sustentam a programação orientada a objetos, que tem estas características definidoras:

-   A maior parte do cálculo é expressa em termos de operações sobre objetos.

-   Os objetos muitas vezes representam coisas do mundo real, e os métodos muitas vezes correspondem aos jeitos em que as coisas do mundo real interagem.

-   Os programas incluem definições de classe e de método.

Por exemplo, no capítulo anterior definimos uma classe `Time` que corresponde ao jeito como as pessoas registram o horário do dia, e definimos funções que correspondem aos tipos de coisa que as pessoas fazem com horários.
Mas não havia conexão explícita entre a definição da classe `Time` e as definições de função que se seguem.
Podemos tornar a conexão explícita reescrevendo uma função como um **método**, que é definido dentro de uma definição de classe.

## O método `__str__`

Quando você escreve um método, pode escolher quase qualquer nome que quiser.
No entanto, alguns nomes têm significados especiais.
Por exemplo, se um objeto tem um método chamado `__str__`, o Python usa esse método para converter o objeto em uma string.
Por exemplo, aqui está um método `__str__` para um objeto time.

Este método é semelhante a `print_time`, do capítulo anterior, com a diferença de que devolve a string em vez de imprimi-la.

Você pode invocar este método do jeito usual.

Mas o Python também pode invocá-lo por você.
Se você usa a função nativa `str` para converter um objeto `Time` em string, o Python usa o método `__str__` da classe `Time`.

E faz o mesmo se você imprimir um objeto `Time`.

Métodos como `__str__` se chamam **métodos especiais**.
Você pode identificá-los porque os nomes começam e terminam com dois underscores.

## Sobrecarga de operadores

Definindo outros métodos especiais, você pode especificar o comportamento de
operadores em tipos definidos pelo programador. Por exemplo, se você definir um
método chamado `__add__` para a classe `Time`, pode usar o operador `+`
em objetos Time.

Há muita coisa acontecendo quando rodamos essas três linhas de código:

* Quando instanciamos um objeto `Time`, o método `__init__` é invocado.

* Quando usamos o operador `+` com um objeto `Time`, o método `__add__` é invocado.

* E quando imprimimos um objeto `Time`, o método `__str__` é invocado.

Mudar o comportamento de um operador para que funcione com tipos definidos pelo programador se chama **sobrecarga de operadores**.
Para cada operador, como `+`, há um método especial correspondente, como `__add__`. 

## Depuração

Um objeto `Time` é válido se os valores de `minute` e `second` estiverem entre `0` e `60`, incluindo `0` mas não `60`, e se `hour` for positivo.
Além disso, `hour` e `minute` devem ser valores inteiros, mas podemos permitir que `second` tenha uma parte fracionária.
Requisitos como estes se chamam **invariantes** porque devem ser sempre verdadeiros.
Dito de outro modo, se não forem verdadeiros, alguma coisa deu errado.

Escrever código para verificar invariantes pode ajudar a detectar erros e encontrar suas causas.
Por exemplo, você pode ter um método como `is_valid` que recebe um objeto Time e devolve `False` se ele viola um invariante.

Então, no início de cada método, você pode verificar os argumentos para ter certeza de que são válidos.

A instrução `assert` avalia a expressão que se segue. Se o resultado for `True`, não faz nada; se o resultado for `False`, causa um `AssertionError`.

## Glossário

**linguagem orientada a objetos:**
Uma linguagem que oferece recursos para sustentar a programação orientada a objetos, sobretudo tipos definidos pelo usuário.

**método:**
Uma função que é definida dentro de uma definição de classe e é invocada sobre instâncias dessa classe.

**receptor:**
O objeto sobre o qual um método é invocado.

**método estático:**
Um método que pode ser invocado sem um objeto como receptor.

**método de instância:**
Um método que precisa ser invocado com um objeto como receptor.

**método especial:**
Um método que muda o jeito como operadores e algumas funções trabalham com um objeto.

**sobrecarga de operadores:**
O processo de usar métodos especiais para mudar o jeito como operadores trabalham com tipos definidos pelo usuário.

**invariante:**
 Uma condição que deve ser sempre verdadeira durante a execução de um programa.
