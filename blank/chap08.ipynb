# Strings e expressões regulares

Strings não são como inteiros, floats e booleanos. Uma string é uma **sequência**, o que significa que contém vários valores em uma ordem particular.
Neste capítulo vamos ver como acessar os valores que formam uma string, e vamos usar funções que processam strings.

Também vamos usar expressões regulares, que são uma ferramenta poderosa para achar padrões em uma string e realizar operações como busca e substituição.

Como exercício, você vai ter a chance de aplicar essas ferramentas a um jogo de palavras chamado Wordle.

## Uma string é uma sequência

Uma string é uma sequência de caracteres. Um **caractere** pode ser uma letra (em quase qualquer alfabeto), um dígito, um sinal de pontuação ou espaço em branco.

Você pode selecionar um caractere de uma string com o operador de colchetes.

A expressão entre colchetes é um **índice**, assim chamado porque *indica* qual caractere da sequência selecionar.
Mas o resultado talvez não seja o que você espera.

A letra com índice `1` é, na verdade, a segunda letra da string.
Um índice é um deslocamento a partir do início da string, então o deslocamento da primeira letra é `0`.

Você pode pensar em `'b'` como a 0ª letra de `'banana'`, pronunciada "zero-ésima".

O índice entre colchetes pode ser uma variável.

## Métodos de string

Strings oferecem métodos que realizam uma variedade de operações úteis. 
Um método é parecido com uma função: recebe argumentos e devolve um valor, mas a sintaxe é diferente.
Por exemplo, o método `upper` recebe uma string e devolve uma string nova com todas as letras maiúsculas.

Em vez da sintaxe de função `upper(word)`, ele usa a sintaxe de método `word.upper()`.

Este uso do operador ponto especifica o nome do método, `upper`, e o nome da string à qual aplicar o método, `word`.
Os parênteses vazios indicam que este método não recebe argumentos.

Uma chamada de método se chama **invocação**; neste caso, diríamos que estamos invocando `upper` em `word`.

Se estiver trabalhando em um notebook Jupyter, pode usar **comandos de shell** para ajudar.
Por exemplo, para exibir as primeiras linhas de um arquivo, você pode usar o comando `!head`, assim:

O ponto de exclamação inicial, `!`, indica que isto é um comando de shell, que não faz parte de Python.
Para exibir as últimas linhas, você pode usar `!tail`.

Quando você trabalha com arquivos grandes, a depuração pode ser difícil porque pode haver saída demais para conferir à mão.

Para fazer um arquivo pequeno que contém parte de um arquivo maior, podemos usar `!head` de novo com o operador de redirecionamento, `>`, que indica que os resultados devem ser escritos em um arquivo em vez de exibidos.

## Glossário

**sequência:**
 Uma coleção ordenada de valores em que cada valor é identificado por um índice inteiro.

**caractere:**
Um elemento de uma string, inclusive letras, números e símbolos.

**índice:**
 Um valor inteiro usado para selecionar um item em uma sequência, como um caractere em uma string. Em Python os índices começam em `0`.

**fatia:**
 Uma parte de uma string especificada por um intervalo de índices.

**string vazia:**
Uma string que não contém nenhum caractere e tem comprimento `0`.

**objeto:**
 Algo a que uma variável pode se referir. Um objeto tem um tipo e um valor.

**imutável:**
Se os elementos de um objeto não podem ser mudados, o objeto é imutável.

**invocação:**
 Uma expressão, ou parte de uma expressão, que chama um método.

**expressão regular:**
Uma sequência de caracteres que define um padrão de busca.

**padrão:**
Uma regra que especifica os requisitos que uma string precisa cumprir para constituir uma correspondência.

**substituição de string:**
Substituição de uma string, ou de parte de uma string, por outra string.

**comando de shell:**
Uma instrução em uma linguagem de shell, que é uma linguagem usada para interagir com um sistema operacional.
