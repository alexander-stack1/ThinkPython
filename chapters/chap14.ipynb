## Tipos definidos pelo programador

Já usamos muitos tipos nativos de Python. Agora vamos definir um tipo novo.
Como primeiro exemplo, vamos criar um tipo chamado `Time` que representa um horário do dia.
Um tipo definido pelo programador também se chama **classe**.
Uma definição de classe fica assim:

```python
class Time:
    """Represents a time of day."""
```

O cabeçalho indica que a classe nova se chama `Time`.
O corpo é uma docstring que explica para que serve a classe.
Definir uma classe cria um **objeto de classe**.

O objeto de classe é como uma fábrica para criar objetos.
Para criar um objeto `Time`, você chama `Time` como se fosse uma função.

```python
lunch = Time()
```

O resultado é um objeto novo cujo tipo é `__main__.Time`, em que `__main__` é o nome do módulo em que `Time` está definido.

```python
type(lunch)
```

Quando você imprime um objeto, o Python diz qual é o tipo e onde ele está guardado na memória (o prefixo `0x` significa que o número seguinte está em hexadecimal).

```python
print(lunch)
```

Criar um objeto novo se chama **instanciação**, e o objeto é uma **instância** da classe.

## Atributos

Um objeto pode conter variáveis, que se chamam **atributos** e se pronunciam com ênfase na primeira sílaba, como "AT-tri-bu-to", e não na segunda, como "a-TRI-bu-to".
Podemos criar atributos usando notação de ponto.

```python
lunch.hour = 11
lunch.minute = 59
lunch.second = 1
```

Este exemplo cria atributos chamados `hour`, `minute` e `second`, que contêm as horas, os minutos e os segundos do horário `11:59:01`, que é hora do almoço, pelo menos para mim.

O diagrama a seguir mostra o estado de `lunch` e de seus atributos depois dessas atribuições.

A variável `lunch` se refere a um objeto `Time`, que contém três atributos.

Um diagrama de estado como este, que mostra um objeto e seus atributos, se chama **diagrama de objeto**.

Você pode ler o valor de um atributo usando o operador de ponto.

```python
lunch.hour
```

Você pode usar um atributo como parte de qualquer expressão.

```python
total_minutes = lunch.hour * 60 + lunch.minute
total_minutes
```

E pode usar o operador de ponto em uma expressão em uma f-string.

```python
f'{lunch.hour}:{lunch.minute}:{lunch.second}'
```

Mas observe que o exemplo anterior não está no formato padrão.
Para corrigir, temos de imprimir os atributos `minute` e `second` com um zero à esquerda.
Podemos fazer isso estendendo as expressões entre chaves com um **especificador de formato**.
No exemplo a seguir, os especificadores de formato indicam que `minute` e `second` devem ser exibidos com pelo menos dois dígitos e um zero à esquerda se necessário.

```python
f'{lunch.hour}:{lunch.minute:02d}:{lunch.second:02d}'
```

Vamos usar esta f-string para escrever uma função que exibe o valor de um objeto `Time`.
Você pode passar um objeto como argumento do jeito usual.
Por exemplo, a função a seguir recebe um objeto `Time` como argumento. 

```python
def print_time(time):
    s = f'{time.hour:02d}:{time.minute:02d}:{time.second:02d}'
    print(s)
```

Quando a chamamos, podemos passar `lunch` como argumento.

```python
print_time(lunch)
```

## Objetos como valores de retorno

As funções podem devolver objetos. Por exemplo, `make_time` recebe parâmetros chamados `hour`, `minute` e `second`, guarda-os como atributos em um objeto `Time` e devolve o objeto novo.

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

Pode surpreender que os parâmetros tenham os mesmos nomes dos atributos, mas esse é um jeito comum de escrever uma função como esta.
Eis como usamos `make_time` para criar um objeto `Time`.

```python
time = make_time(11, 59, 1)
print_time(time)
```

## Objetos são mutáveis

Suponha que você vai a uma sessão de um filme, como *Monty Python and the Holy Grail*, que começa às `21:20` e dura `92` minutos, isto é, uma hora e `32` minutos.
A que horas o filme termina?

Primeiro, vamos criar um objeto `Time` que representa o horário de início.

```python
start = make_time(9, 20, 0)
print_time(start)
```

Para encontrar o horário de término, podemos modificar os atributos do objeto `Time`, somando a duração do filme.

```python
start.hour += 1
start.minute += 32
print_time(start)
```

O filme vai terminar às 22:52.

Vamos encapsular este cálculo em uma função e generalizá-lo para receber a duração do filme em três parâmetros: `hours`, `minutes` e `seconds`.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
```

Aqui está um exemplo que demonstra o efeito.

```python
start = make_time(9, 20, 0)
increment_time(start, 1, 32, 0)
print_time(start)
```

Dentro da função, `time` é um alias de `start`, então quando `time` é modificado, `start` muda.

Esta função funciona, mas depois que ela roda, ficamos com uma variável chamada `start` que se refere a um objeto que representa o horário de *término*, e não temos mais um objeto que representa o horário de início.
Seria melhor deixar `start` inalterado e criar um objeto novo para representar o horário de término.
Podemos fazer isso copiando `start` e modificando a cópia.

## Cópia

O módulo `copy` oferece uma função chamada `copy` que pode duplicar qualquer objeto.
Podemos importá-lo assim.

```python
from copy import copy
```

Para ver como funciona, vamos começar com um objeto `Time` novo que representa o horário de início do filme.

```python
start = make_time(9, 20, 0)
```

E fazer uma cópia.

```python
end = copy(start)
```

Agora `start` e `end` contêm os mesmos dados.

```python
print_time(start)
print_time(end)
```

Mas o operador `is` confirma que não são o mesmo objeto.

```python
start is end
```

Vamos ver o que o operador `==` faz.

```python
start == end
```

Você poderia esperar que `==` produzisse `True` porque os objetos contêm os mesmos dados.
Mas para classes definidas pelo programador, o comportamento padrão do operador `==` é o mesmo do operador `is`: ele verifica identidade, não equivalência.

## Funções puras

Podemos usar `copy` para escrever funções puras que não modificam seus parâmetros.
Por exemplo, aqui está uma função que recebe um objeto `Time` e uma duração em horas, minutos e segundos.
Ela faz uma cópia do objeto original, usa `increment_time` para modificar a cópia e a devolve.

```python
def add_time(time, hours, minutes, seconds):
    total = copy(time)
    increment_time(total, hours, minutes, seconds)
    return total
```

Eis como a usamos.

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

O valor de retorno é um objeto novo que representa o horário de término do filme.
E podemos confirmar que `start` não mudou.

```python
print_time(start)
```

`add_time` é uma **função pura** porque não modifica nenhum dos objetos passados a ela como argumentos e seu único efeito é devolver um valor.

Tudo que pode ser feito com funções impuras também pode ser feito com funções puras.
De fato, algumas linguagens de programação só permitem funções puras.
Programas que usam funções puras podem ser menos propensos a erros, mas funções impuras às vezes são convenientes e podem ser mais eficientes.

Em geral, sugiro que você escreva funções puras sempre que for razoável e recorra a funções impuras só se houver uma vantagem convincente.
Essa abordagem pode ser chamada de **estilo de programação funcional**.

## Protótipo e remendo

No exemplo anterior, `increment_time` e `add_time` parecem funcionar, mas se tentarmos outro exemplo, veremos que não estão bem corretas.

Suponha que você chega ao cinema e descobre que o filme começa às `21:40`, e não às `21:20`.
Eis o que acontece quando calculamos o horário de término atualizado.

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

O resultado não é um horário válido.
O problema é que `increment_time` não lida com casos em que o número de segundos ou minutos soma mais de `60`.

Aqui está uma versão melhorada que verifica se `second` é maior ou igual a `60`: se for, incrementa `minute`; depois verifica se `minute` é maior ou igual a `60`: se for, incrementa `hour`.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds

    if time.second >= 60:
        time.second -= 60
        time.minute += 1

    if time.minute >= 60:
        time.minute -= 60
        time.hour += 1
```

Corrigir `increment_time` também corrige `add_time`, que a usa.
Assim, agora o exemplo anterior funciona corretamente.

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

Mas esta função ainda não está correta, porque os argumentos podem ser maiores que `60`.
Por exemplo, suponha que nos dêem a duração como `92` minutos, em vez de `1` hora e `32` minutos.
Poderíamos chamar `add_time` assim.

```python
end = add_time(start, 0, 92, 0)
print_time(end)
```

O resultado não é um horário válido.
Então vamos tentar outra abordagem, usando a função `divmod`.
Vamos fazer uma cópia de `start` e modificá-la incrementando o atributo `minute`.

```python
end = copy(start)
end.minute = start.minute + 92
end.minute
```

Agora `minute` é `132`, que é `2` horas e `12` minutos.
Podemos usar `divmod` para dividir por `60` e devolver o número de horas inteiras e o número de minutos restantes.

```python
carry, end.minute = divmod(end.minute, 60)
carry, end.minute
```

Agora `minute` está correto, e podemos somar as horas a `hour`.

```python
end.hour += carry
print_time(end)
```

O resultado é um horário válido.
Podemos fazer o mesmo com `hour` e `second` e encapsular o processo inteiro em uma função.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
    
    carry, time.second = divmod(time.second, 60)
    carry, time.minute = divmod(time.minute + carry, 60)
    carry, time.hour = divmod(time.hour + carry, 24)
```

Com esta versão de `increment_time`, `add_time` funciona corretamente, mesmo se os argumentos ultrapassarem `60`.

```python
end = add_time(start, 0, 90, 120)
print_time(end)
```

Esta seção demonstra um plano de desenvolvimento de programas que chamo de **protótipo e remendo**.
Começamos com um protótipo simples que funcionava corretamente no primeiro exemplo.
Depois o testamos com exemplos mais difíceis: quando encontramos um erro, modificamos o programa para corrigi-lo, como se colocássemos um remendo em um pneu furado.

Essa abordagem pode ser eficaz, sobretudo se você ainda não tem um entendimento profundo do problema.
Mas correções incrementais podem gerar código desnecessariamente complicado, já que lida com muitos casos especiais, e pouco confiável, já que é difícil saber se você
encontrou todos os erros.

## Desenvolvimento pelo projeto

Um plano alternativo é o **desenvolvimento pelo projeto**, que envolve mais planejamento antes de prototipar. Em um processo centrado no projeto, às vezes uma percepção de alto nível do problema torna a programação bem mais fácil.

Neste caso, a percepção é que podemos pensar em um objeto `Time` como um número de três dígitos na base 60, também conhecida como sexagesimal.
O atributo `second` é a "coluna das unidades", o atributo `minute` é a "coluna das sessentas"
e o atributo `hour` é a "coluna das três mil e seiscentas".
Quando escrevemos `increment_time`, estávamos efetivamente fazendo adição na base 60, e por isso tínhamos de transportar de uma coluna para a seguinte.

Essa observação sugere outra abordagem para o problema inteiro: podemos converter objetos `Time` em inteiros e aproveitar o fato de que o Python sabe fazer aritmética de inteiros.

Aqui está uma função que converte de um `Time` para um inteiro.

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

O resultado é o número de segundos desde o início do dia.
Por exemplo, `01:01:01` é `1` hora, `1` minuto e `1` segundo desde o início do dia, que é a soma de `3600` segundos, `60` segundos e `1` segundo.

```python
time = make_time(1, 1, 1)
print_time(time)
time_to_int(time)
```

E aqui está uma função que vai na outra direção, convertendo um inteiro em um objeto `Time`, usando a função `divmod`.

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

Podemos testá-la convertendo o exemplo anterior de volta em um `Time`.

```python
time = int_to_time(3661)
print_time(time)
```

Usando essas funções, podemos escrever uma versão mais concisa de `add_time`.

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

A primeira linha converte os argumentos em um objeto `Time` chamado `duration`.
A segunda linha converte `time` e `duration` em segundos e os soma.
A terceira linha converte a soma em um objeto `Time` e o devolve.

Eis como funciona.

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

Em alguns aspectos, converter da base 60 para a base 10 e de volta é mais difícil do que
só lidar com horários. Conversão de base é mais abstrata; nossa intuição
para lidar com valores de tempo é melhor.

Mas se tivermos a percepção de tratar horários como números na base 60, e investirmos o esforço de escrever as funções de conversão `time_to_int` e `int_to_time`, obtemos um programa mais curto, mais fácil de ler e depurar, e mais confiável.

Também fica mais fácil acrescentar recursos depois. Por exemplo, imagine subtrair dois objetos `Time` para encontrar a duração entre eles.
A abordagem ingênua é implementar a subtração com empréstimo.
Usar as funções de conversão é mais fácil e tem mais chance de estar correto.

Por ironia, às vezes tornar um problema mais difícil, ou mais geral, o torna mais fácil, porque há menos casos especiais e menos oportunidades de erro.

## Glossário

**programação orientada a objetos:**
Um estilo de programação que usa objetos para organizar código e dados.

**classe:**
 Um tipo definido pelo programador. Uma definição de classe cria um objeto de classe novo.

**objeto de classe:**
Um objeto que representa uma classe: é o resultado de uma definição de classe.

**instanciação:**
O processo de criar um objeto que pertence a uma classe.

**instância:**
 Um objeto que pertence a uma classe.

**atributo:**
 Uma variável associada a um objeto, também chamada de variável de instância.

**diagrama de objeto:**
Uma representação gráfica de um objeto, de seus atributos e de seus valores.

**especificador de formato:**
Em uma f-string, um especificador de formato determina como um valor é convertido em string.

**função pura:**
Uma função que não modifica seus parâmetros nem tem qualquer efeito além de devolver um valor.

**estilo de programação funcional:**
Um jeito de programar que usa funções puras sempre que possível.

**protótipo e remendo:**
Um jeito de desenvolver programas começando com um rascunho e acrescentando recursos e corrigindo bugs aos poucos.

**desenvolvimento pelo projeto:**
Um jeito de desenvolver programas com planejamento mais cuidadoso do que protótipo e remendo.
