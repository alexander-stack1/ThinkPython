# Prefácio



## Para quem é este livro?

Se você quer aprender a programar, veio ao lugar certo.
Python é uma das melhores linguagens para quem está começando e também uma das habilidades mais procuradas.

Você também chegou em um bom momento, porque aprender a programar hoje provavelmente é mais fácil do que nunca.
Com assistentes virtuais como o ChatGPT, você não precisa aprender sozinho.
Ao longo deste livro, sugiro formas de usar essas ferramentas para acelerar o aprendizado.

Este livro é sobretudo para quem nunca programou e para quem já tem alguma experiência em outra linguagem.
Se você já tem bastante experiência em Python, pode achar os primeiros capítulos lentos demais.

Um dos desafios de aprender a programar é que você precisa aprender *duas* linguagens: uma é a própria linguagem de programação; a outra é o vocabulário que usamos para falar sobre programas.
Se você aprender só a linguagem de programação, provavelmente terá dificuldade para interpretar uma mensagem de erro, ler documentação, conversar com outra pessoa ou usar assistentes virtuais.
Se você já programou, mas ainda não aprendeu essa segunda linguagem, espero que este livro seja útil.

## Objetivos do livro

Ao escrever este livro, procurei ser cuidadoso com o vocabulário.
Defino cada termo na primeira vez em que ele aparece.
E há um glossário no fim de cada capítulo, que revisa os termos introduzidos.

Também procurei ser conciso.
Quanto menor o esforço mental para ler o livro, maior a capacidade que sobra para programar.

Mas ninguém aprende a programar só lendo um livro: é preciso praticar.
Por isso, este livro inclui exercícios no fim de cada capítulo, para você praticar o que aprendeu.

Se você ler com atenção e fizer os exercícios com constância, vai avançar.
Mas já aviso: aprender a programar não é fácil, e mesmo programadores experientes se frustram.
Ao longo do caminho, sugiro estratégias para escrever programas corretos e corrigir os incorretos.

## Como navegar pelo livro

Cada capítulo deste livro se apoia nos anteriores, então leia na ordem e reserve tempo para os exercícios antes de seguir adiante.

Os seis primeiros capítulos apresentam elementos básicos como aritmética, condicionais e laços.
Também apresentam o conceito mais importante da programação, as funções, e um modo poderoso de usá-las, a recursão.

Os capítulos 7 e 8 apresentam as strings, que podem representar letras, palavras e frases, e algoritmos para trabalhar com elas.

Os capítulos 9 a 12 apresentam as estruturas de dados centrais de Python: listas, dicionários e tuplas, ferramentas poderosas para escrever programas eficientes.
O capítulo 12 apresenta algoritmos para analisar texto e gerar texto novo de forma aleatória.
Algoritmos assim estão no núcleo dos grandes modelos de linguagem (LLMs), então este capítulo dá uma ideia de como funcionam ferramentas como o ChatGPT.

O capítulo 13 trata de formas de guardar dados em armazenamento de longo prazo: arquivos e bancos de dados.
Como exercício, você pode escrever um programa que percorre um sistema de arquivos e encontra arquivos duplicados.

Os capítulos 14 a 17 apresentam a programação orientada a objetos (POO), um modo de organizar programas e os dados com os quais eles trabalham.
Muitas bibliotecas de Python são escritas nesse estilo, então esses capítulos ajudam a entender o projeto delas e a definir os seus próprios objetos.

O objetivo deste livro não é cobrir a linguagem Python inteira.
Em vez disso, concentro-me em um subconjunto que oferece a maior capacidade com o menor número de conceitos.
Ainda assim, Python tem muitos recursos que resolvem problemas comuns com eficiência.
O capítulo 18 apresenta alguns deles.

Por fim, o capítulo 19 traz minhas considerações finais e sugestões para continuar a jornada na programação.

## O que há de novo na terceira edição?

As maiores mudanças desta edição foram impulsionadas por duas tecnologias novas: notebooks Jupyter e assistentes virtuais.

Cada capítulo deste livro é um notebook Jupyter, um documento que contém texto comum e código.
Para mim, isso facilita escrever o código, testá-lo e mantê-lo consistente com o texto.
Para você, significa que pode executar o código, modificá-lo e fazer os exercícios, tudo no mesmo lugar.
As instruções para trabalhar com os notebooks estão no primeiro capítulo.

A outra grande mudança é que incluí conselhos para trabalhar com assistentes virtuais como o ChatGPT e usá-los para acelerar o aprendizado.
Quando a edição anterior deste livro foi publicada, em 2016, os predecessores dessas ferramentas eram bem menos úteis e a maior parte das pessoas nem as conhecia.
Hoje elas são ferramentas padrão da engenharia de software, e creio que serão transformadoras para aprender a programar e para aprender muita coisa além disso.

As demais mudanças do livro nasceram dos meus arrependimentos em relação à segunda edição.

O primeiro é que não dei ênfase a testes de software.
Isso já era uma omissão lamentável em 2016, mas com o surgimento dos assistentes virtuais o teste automatizado ficou ainda mais importante.
Por isso, esta edição apresenta as ferramentas de teste mais usadas em Python, `doctest` e `unittest`, e inclui vários exercícios para praticá-las.

O outro arrependimento é que os exercícios da segunda edição eram desiguais: alguns eram mais interessantes do que outros, e alguns eram difíceis demais.
Passar para notebooks Jupyter me ajudou a desenvolver e testar uma sequência de exercícios mais envolvente e eficaz.

Nesta revisão, a sequência de tópicos é quase a mesma, mas reorganizei alguns capítulos e comprimi dois capítulos curtos em um só.
Também ampliei o tratamento de strings para incluir expressões regulares.

Alguns capítulos usam gráficos de tartaruga.
Nas edições anteriores, usei o módulo `turtle` de Python, mas infelizmente ele não funciona em notebooks Jupyter.
Por isso, substituí-o por um módulo novo de tartaruga, que deve ser mais fácil de usar.

Por fim, reescrevi uma fração substancial do texto, esclarecendo trechos que precisavam disso e enxugando pontos em que eu não era tão conciso quanto poderia.

Tenho muito orgulho desta edição nova. Espero que você goste!

## Como começar

Para a maior parte das linguagens de programação, inclusive Python, existem muitas ferramentas para escrever e executar programas.
Essas ferramentas se chamam ambientes de desenvolvimento integrado (IDEs).
Em geral, há dois tipos de IDE:

* Algumas trabalham com arquivos que contêm código, e por isso oferecem ferramentas para editar e executar esses arquivos.

* Outras trabalham sobretudo com notebooks, que são documentos com texto e código.

Para quem está começando, recomendo um ambiente de notebooks como o Jupyter.

Os notebooks deste livro estão em um repositório on-line em <https://allendowney.github.io/ThinkPython>.

Há duas formas de usá-los:

* Você pode baixar os notebooks e executá-los no seu computador. Nesse caso, precisa instalar Python e Jupyter, o que não é difícil, mas se o seu objetivo é aprender Python pode ser frustrante gastar muito tempo instalando software.

* A alternativa é executar os notebooks no Colab, um ambiente Jupyter que roda no navegador, sem instalar nada. O Colab é operado pelo Google e é gratuito.

Se você está começando agora, recomendo fortemente começar pelo Colab.

## Recursos para quem ensina

Se você vai ensinar com este livro, aqui estão alguns recursos que podem ser úteis.

* Há notebooks com soluções dos exercícios em <https://allendowney.github.io/ThinkPython>, além de links para os recursos abaixo.

* Questionários de cada capítulo, e um questionário somativo do livro inteiro, estão disponíveis sob consulta.

* *Teaching and Learning with Jupyter* é um livro on-line com sugestões para usar o Jupyter com eficácia em sala de aula. Você pode lê-lo em <https://jupyter4edu.github.io/jupyter-edu-book>

* Uma das melhores formas de usar notebooks é a programação ao vivo (*live coding*), em que a pessoa que ensina escreve código e a turma acompanha nos próprios notebooks. Para aprender sobre programação ao vivo e receber outros bons conselhos sobre ensino de programação, recomendo a formação de instrutores da The Carpentries, em <https://carpentries.github.io/instructor-training>
