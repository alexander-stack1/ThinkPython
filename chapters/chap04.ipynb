# Funções e interfaces

Este capítulo apresenta um módulo chamado `jupyturtle`, que permite criar desenhos simples dando instruções a uma tartaruga imaginária.
Vamos usar este módulo para escrever funções que desenham quadrados, polígonos e círculos, e para demonstrar o **projeto de interface**, um jeito de projetar funções que trabalham juntas.

## O módulo jupyturtle

Para usar o módulo `jupyturtle`, podemos importá-lo assim.

````python
import jupyturtle
````

Agora podemos usar as funções definidas no módulo, como `make_turtle` e `forward`.

````python
jupyturtle.make_turtle()
jupyturtle.forward(100)
````

`make_turtle` cria um **canvas**, que é um espaço na tela onde podemos desenhar, e uma tartaruga, representada por um casco circular e uma cabeça triangular.
O círculo mostra a localização da tartaruga e o triângulo indica a direção para a qual ela está virada.

`forward` move a tartaruga uma distância dada na direção para a qual ela está virada, desenhando um segmento de reta pelo caminho.
A distância está em unidades arbitrárias: o tamanho real depende da tela do seu computador.

Vamos usar funções definidas no módulo `jupyturtle` muitas vezes, então seria bom não ter de escrever o nome do módulo toda vez.
Isso é possível se importarmos o módulo assim.

````python
from jupyturtle import make_turtle, forward
````

Esta versão da instrução import importa `make_turtle` e `forward` do módulo `jupyturtle` para que possamos chamá-las assim.

````python
make_turtle()
forward(100)
````

`jupyturtle` oferece duas outras funções que vamos usar, chamadas `left` e `right`.
Vamos importá-las assim.

````python
from jupyturtle import left, right
````

`left` faz a tartaruga virar à esquerda. Ela recebe um argumento, que é o ângulo da virada em graus.
Por exemplo, podemos fazer uma virada de 90 graus à esquerda assim.

````python
make_turtle()
forward(50)
left(90)
forward(50)
````

Este programa move a tartaruga para o leste e depois para o norte, deixando dois segmentos de reta para trás.
Antes de seguir, veja se consegue modificar o programa anterior para fazer um quadrado.

## Fazendo um quadrado

Aqui está um jeito de fazer um quadrado.

````python
make_turtle()

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)
````

Como este programa repete o mesmo par de linhas quatro vezes, podemos fazer a mesma coisa de forma mais concisa com um laço `for`.

````python
make_turtle()
for i in range(4):
    forward(50)
    left(90)
````

## Encapsulamento e generalização

Vamos pegar o código que desenha o quadrado da seção anterior e colocá-lo numa função chamada `square`.

````python
def square():
    for i in range(4):
        forward(50)
        left(90)
````

Agora podemos chamar a função assim.

````python
make_turtle()
square()
````

Embrulhar um pedaço de código numa função se chama **encapsulamento**.
Um dos benefícios do encapsulamento é que ele associa um nome ao código, o que serve como uma espécie de documentação. Outra vantagem é que, se você reutilizar o código, é mais conciso chamar uma função duas vezes do que copiar e colar o corpo!

Na versão atual, o tamanho do quadrado é sempre `50`.
Se quisermos desenhar quadrados de tamanhos diferentes, podemos receber o comprimento dos lados como parâmetro.

````python
def square(length):
    for i in range(4):
        forward(length)
        left(90)
````

Agora podemos desenhar quadrados de tamanhos diferentes.

````python
make_turtle()
square(30)
square(60)
````

Acrescentar um parâmetro a uma função se chama **generalização** porque deixa a função mais geral: na versão anterior, o quadrado tem sempre o mesmo tamanho; nesta versão, pode ter qualquer tamanho.

Se acrescentarmos outro parâmetro, podemos torná-la ainda mais geral.
A função a seguir desenha polígonos regulares com uma quantidade dada de lados.

````python
def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)
````

Num polígono regular com `n` lados, o ângulo entre lados adjacentes é `360 / n` graus.

O exemplo a seguir desenha um polígono de `7` lados com comprimento de lado `30`.

````python
make_turtle()
polygon(7, 30)
````

Quando uma função tem mais do que uns poucos argumentos numéricos, é fácil esquecer o que eles são, ou em que ordem devem estar.
Pode ser uma boa ideia incluir os nomes dos parâmetros na lista de argumentos.

````python
make_turtle()
polygon(n=7, length=30)
````

Às vezes esses se chamam "argumentos nomeados" porque incluem os nomes dos parâmetros.
Mas em Python eles são mais frequentemente chamados de **argumentos de palavra-chave** (não confundir com as palavras-chave de Python como `for` e `def`).

Este uso do operador de atribuição, `=`, é um lembrete de como argumentos e parâmetros funcionam: quando você chama uma função, os argumentos são atribuídos aos parâmetros.

## Aproximando um círculo

Agora suponha que queremos desenhar um círculo.
Podemos fazer isso, aproximadamente, desenhando um polígono com um número grande de lados, de modo que cada lado seja pequeno o bastante para ficar difícil de ver.
Aqui está uma função que usa `polygon` para desenhar um polígono de `30` lados que aproxima um círculo.

````python
import math

def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

`circle` recebe o raio do círculo como parâmetro.
Ela calcula `circumference`, que é a circunferência de um círculo com o raio dado.
`n` é o número de lados, então `circumference / n` é o comprimento de cada lado.

Esta função pode demorar para executar.
Podemos acelerá-la chamando `make_turtle` com um argumento de palavra-chave chamado `delay` que define o tempo, em segundos, que a tartaruga espera depois de cada passo.
O valor padrão é `0.2` segundos: se definirmos `0.02`, ela roda cerca de 10 vezes mais depressa.

````python
make_turtle(delay=0.02)
circle(30)
````

Uma limitação desta solução é que `n` é uma constante, o que significa
que, para círculos muito grandes, os lados ficam longos demais, e para círculos
pequenos, perdemos tempo desenhando lados muito curtos.
Uma opção é generalizar a função recebendo `n` como parâmetro.
Mas vamos manter as coisas simples por enquanto.

## Refatoração

Agora vamos escrever uma versão mais geral de `circle`, chamada `arc`, que recebe um segundo parâmetro, `angle`, e desenha um arco de círculo que cobre o ângulo dado.
Por exemplo, se `angle` for `360` graus, ela desenha um círculo completo. Se `angle` for `180` graus, desenha um semicírculo.

Para escrever `circle`, pudemos reutilizar `polygon`, porque um polígono de muitos lados é uma boa aproximação de um círculo.
Mas não podemos usar `polygon` para escrever `arc`.

Em vez disso, vamos criar a versão mais geral de `polygon`, chamada `polyline`.

````python
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)
````

`polyline` recebe como parâmetros o número de segmentos de reta a desenhar, `n`, o comprimento dos segmentos, `length`, e o ângulo entre eles, `angle`.

````python
def polygon(n, length):
    angle = 360.0 / n
    polyline(n, length, angle)
````

E podemos usar `polyline` para escrever `arc`.

````python
def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)
````

`arc` é semelhante a `circle`, exceto que calcula `arc_length`, que é uma fração da circunferência de um círculo.

Por fim, podemos reescrever `circle` para usar `arc`.

````python
def circle(radius):
    arc(radius,  360)
````

Para conferir se essas funções funcionam como esperado, vamos usá-las para desenhar algo parecido com um caracol.
Com `delay=0`, a tartaruga corre o mais depressa possível.

````python
make_turtle(delay=0)
polygon(n=20, length=9)
arc(radius=70, angle=70)
circle(radius=10)
````

Neste exemplo, começamos com código que funcionava e o reorganizamos com funções diferentes.
Mudanças como esta, que melhoram o código sem mudar o comportamento, se chamam **refatoração**.

Se tivéssemos planejado com antecedência, talvez tivéssemos escrito `polyline` primeiro e evitado a refatoração, mas muitas vezes você não sabe o bastante no começo de um projeto para projetar todas as funções.
Quando começa a programar, você entende melhor o problema.
Às vezes a refatoração é um sinal de que você aprendeu alguma coisa.

## Diagrama de pilha

Quando chamamos `circle`, ela chama `arc`, que chama `polyline`.
Podemos usar um diagrama de pilha para mostrar esta sequência de chamadas de função e os parâmetros de cada uma.

````python
from diagram import make_binding, make_frame, Frame, Stack

frame1 = make_frame(dict(radius=30), name='circle', loc='left')

frame2 = make_frame(dict(radius=30, angle=360), name='arc', loc='left', dx=1.1)

frame3 = make_frame(dict(n=60, length=3.04, angle=5.8), 
                    name='polyline', loc='left', dx=1.1, offsetx=-0.27)

stack = Stack([frame1, frame2, frame3], dy=-0.4)
````

Parâmetros são locais, o que significa que você pode usar o mesmo nome de parâmetro em funções diferentes; é uma variável diferente em cada função, e pode se referir a um valor diferente.

## Um plano de desenvolvimento

Um **plano de desenvolvimento** é um processo para escrever programas.
O processo que usamos neste capítulo é "encapsulamento e generalização".
Os passos deste processo são:

1.  Comece escrevendo um programa pequeno, sem definições de função.

2.  Quando o programa estiver funcionando, identifique um pedaço coerente dele,
    encapsule o pedaço numa função e dê um nome a ela.

3.  Generalize a função acrescentando parâmetros adequados.

4.  Repita os passos 1 a 3 até ter um conjunto de funções que funcionam.

5.  Procure oportunidades de melhorar o programa com refatoração. Por
    exemplo, se você tem código semelhante em vários lugares, considere
    extraí-lo para uma função adequadamente geral.

Este processo tem algumas desvantagens: vamos ver alternativas depois, mas pode ser útil se você não souber de antemão como dividir o programa em funções.
Esta abordagem permite projetar conforme avança.

O projeto de uma função tem duas partes:

* A **interface** é como a função é usada, inclusive o nome, os parâmetros que ela recebe e o que a função deve fazer.

* A **implementação** é como a função faz o que deve fazer.

Por exemplo, aqui está a primeira versão de `circle` que escrevemos, que usa `polygon`.

````python
def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

E aqui está a versão refatorada que usa `arc`.

````python
def circle(radius):
    arc(radius,  360)
````

Estas duas funções têm a mesma interface: recebem os mesmos parâmetros e fazem a mesma coisa, mas têm implementações diferentes.

## Docstrings

Uma **docstring** é uma string no começo de uma função que explica a interface ("doc" é abreviação de "documentation").
Aqui está um exemplo:

````python
def polyline(n, length, angle):
    """Draws line segments with the given length and angle between them.
    
    n: integer number of line segments
    length: length of the line segments
    angle: angle between segments (in degrees)
    """    
    for i in range(n):
        forward(length)
        left(angle)
````

Por convenção, docstrings são strings entre aspas triplas, também conhecidas como **strings de várias linhas** porque as aspas triplas permitem que a string ocupe mais de uma linha.

Uma docstring deve:

* Explicar de forma concisa o que a função faz, sem entrar nos detalhes de como funciona,

* Explicar que efeito cada parâmetro tem no comportamento da função, e

* Indicar de que tipo cada parâmetro deve ser, se isso não for óbvio.

Escrever este tipo de documentação é uma parte importante do projeto de interface.
Uma interface bem projetada deve ser simples de explicar; se você tem dificuldade para explicar uma das suas funções, talvez a interface possa ser melhorada.

## Depuração

Uma interface é como um contrato entre uma função e quem a chama. Quem
chama concorda em fornecer certos argumentos e a função concorda em
fazer certo trabalho.

Por exemplo, `polyline` exige três argumentos: `n` precisa ser um inteiro; `length` deve ser um número positivo; e `angle` precisa ser um número, entendido como estando em graus.

Esses requisitos se chamam **pré-condições** porque devem ser verdadeiros antes de a função começar a executar. Por outro lado, as condições no fim da função são **pós-condições**.
As pós-condições incluem o efeito pretendido da função (como desenhar segmentos de reta) e quaisquer efeitos colaterais (como mover a tartaruga ou fazer outras mudanças).

As pré-condições são responsabilidade de quem chama. Se quem chama viola uma pré-condição e a função não funciona corretamente, o bug está em quem chama, não na função.

Se as pré-condições estão satisfeitas e as pós-condições não, o bug está na função. Se as pré-condições e as pós-condições estiverem claras, elas podem ajudar na depuração.

## Glossário

**projeto de interface:**
Um processo para projetar a interface de uma função, inclusive os parâmetros que ela deve receber.

**canvas:**
Uma janela usada para exibir elementos gráficos, inclusive linhas, círculos, retângulos e outras formas.

**encapsulamento:**
 O processo de transformar uma sequência de instruções numa definição de função.

**generalização:**
 O processo de substituir algo desnecessariamente específico (como um número) por algo adequadamente geral (como uma variável ou um parâmetro).

**argumento de palavra-chave:**
Um argumento que inclui o nome do parâmetro.

**refatoração:**
 O processo de modificar um programa que funciona para melhorar as interfaces das funções e outras qualidades do código.

**plano de desenvolvimento:**
Um processo para escrever programas.

**docstring:**
 Uma string que aparece no topo de uma definição de função para documentar a interface da função.

**string de várias linhas:**
Uma string entre aspas triplas que pode ocupar mais de uma linha de um programa.

**pré-condição:**
 Um requisito que deve ser satisfeito por quem chama antes de a função começar.

**pós-condição:**
 Um requisito que deve ser satisfeito pela função antes de ela terminar.

A função a seguir usa `penup` e `pendown` para mover a tartaruga sem deixar rastro.

````python
from jupyturtle import penup, pendown

def jump(length):
    """Move forward length units without leaving a trail.
    
    Postcondition: Leaves the pen down.
    """
    penup()
    forward(length)
    pendown()
````
