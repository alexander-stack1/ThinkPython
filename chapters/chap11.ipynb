## Tuplas são como listas

Uma tupla é uma sequência de valores. Os valores podem ser de qualquer tipo e são indexados por inteiros, então as tuplas se parecem muito com listas.
A diferença importante é que as tuplas são imutáveis.

Para criar uma tupla, você pode escrever uma lista de valores separados por vírgula.

```python
t = 'l', 'u', 'p', 'i', 'n'
type(t)
```

Embora não seja necessário, é comum colocar as tuplas entre parênteses.

```python
t = ('l', 'u', 'p', 'i', 'n')
type(t)
```

Para criar uma tupla com um único elemento, é preciso incluir uma vírgula no final.

```python
t1 = 'p',
type(t1)
```

Um único valor entre parênteses não é uma tupla.

```python
t2 = ('p')
type(t2)
```

Outra forma de criar uma tupla é a função nativa `tuple`. Sem
argumento, ela cria uma tupla vazia.

```python
t = tuple()
t
```

Se o argumento for uma sequência (string, lista ou tupla), o resultado é uma
tupla com os elementos da sequência.

```python
t = tuple('lupin')
t
```

Como `tuple` é o nome de uma função nativa, evite usá-lo como nome de variável.

A maior parte dos operadores de lista também funciona com tuplas.
Por exemplo, o operador de colchetes indexa um elemento.

```python
t[0]
```

E o operador de fatiamento seleciona um intervalo de elementos.

```python
t[1:3]
```

O operador `+` concatena tuplas.

```python
tuple('lup') + ('i', 'n')
```

E o operador `*` duplica uma tupla um número dado de vezes.

```python
tuple('spam') * 2 
```

A função `sorted` funciona com tuplas, mas o resultado é uma lista, não uma tupla.

```python
sorted(t)
```

A função `reversed` também funciona com tuplas.

```python
reversed(t)
```

O resultado é um objeto `reversed`, que podemos converter em lista ou tupla.

```python
tuple(reversed(t))
```

Com base nos exemplos até aqui, pode parecer que as tuplas são iguais às listas.

## Mas as tuplas são imutáveis

Se você tentar modificar uma tupla com o operador de colchetes, recebe um `TypeError`.

```python
%%expect TypeError
t[0] = 'L'
```

E as tuplas não têm nenhum dos métodos que modificam listas, como `append` e `remove`.

```python
%%expect AttributeError

t.remove('l')
```

Lembre que um "atributo" é uma variável ou método associado a um objeto: essa mensagem de erro significa que as tuplas não têm um método chamado `remove`.

Como as tuplas são imutáveis, elas são hashable, o que significa que podem ser usadas como chaves em um dicionário.
Por exemplo, o dicionário a seguir contém duas tuplas como chaves que mapeiam para inteiros.

```python
d = {}
d[1, 2] = 3
d[3, 4] = 7
```

Podemos procurar uma tupla em um dicionário assim:

```python
d[1, 2]
```

Ou, se temos uma variável que se refere a uma tupla, podemos usá-la como chave.

```python
t = (3, 4)
d[t]
```

As tuplas também podem aparecer como valores em um dicionário.

```python
t = tuple('abc')
d = {'key': t}
d
```

## Atribuição de tuplas

Você pode colocar uma tupla de variáveis do lado esquerdo de uma atribuição e uma tupla de valores do lado direito.

```python
a, b = 1, 2
```

Os valores são atribuídos às variáveis da esquerda para a direita: neste exemplo, `a` recebe o valor `1` e `b` recebe o valor `2`.
Podemos exibir os resultados assim:

```python
a, b
```

De modo mais geral, se o lado esquerdo de uma atribuição é uma tupla, o lado direito pode ser qualquer tipo de sequência: string, lista ou tupla. 
Por exemplo, para separar um endereço de e-mail em nome de usuário e domínio, você poderia escrever:

```python
email = 'monty@python.org'
username, domain = email.split('@')
```

O valor de retorno de `split` é uma lista com dois elementos: o primeiro é atribuído a `username`, o segundo a `domain`.

```python
username, domain
```

O número de variáveis à esquerda e o número de valores à
direita precisam ser iguais: caso contrário, você recebe um `ValueError`.

```python
%%expect ValueError
a, b = 1, 2, 3
```

A atribuição de tuplas é útil se você quer trocar os valores de duas variáveis.
Com atribuições convencionais, é preciso usar uma variável temporária, assim:

```python
temp = a
a = b
b = temp
```

Isso funciona, mas com atribuição de tuplas podemos fazer a mesma coisa sem variável temporária.

```python
a, b = b, a
```

Isso funciona porque todas as expressões do lado direito são avaliadas antes de qualquer atribuição.

Também podemos usar atribuição de tuplas em uma instrução `for`.
Por exemplo, para percorrer os itens de um dicionário, podemos usar o método `items`.

```python
d = {'one': 1, 'two': 2}

for item in d.items():
    key, value = item
    print(key, '->', value)
```

A cada passagem pelo laço, `item` recebe uma tupla que contém uma chave e o valor correspondente.

Podemos escrever esse laço de forma mais concisa, assim:

```python
for key, value in d.items():
    print(key, '->', value)
```

A cada passagem pelo laço, uma chave e o valor correspondente são atribuídos diretamente a `key` e `value`.

## Tuplas como valores de retorno

Em sentido estrito, uma função só pode devolver um valor, mas se o
valor for uma tupla, o efeito é o mesmo que devolver vários valores.
Por exemplo, se você quer dividir dois inteiros e calcular o quociente
e o resto, é ineficiente calcular `x//y` e depois `x%y`. É
melhor calcular os dois ao mesmo tempo.

A função nativa `divmod` recebe dois argumentos e devolve uma tupla
de dois valores, o quociente e o resto.

```python
divmod(7, 3)
```

Podemos usar atribuição de tuplas para guardar os elementos da tupla em duas variáveis. 

```python
quotient, remainder = divmod(7, 3)
quotient
```

```python
remainder
```

Aqui está um exemplo de função que devolve uma tupla.

```python
def min_max(t):
    return min(t), max(t)
```

`max` e `min` são funções nativas que encontram o maior e o menor elemento de uma sequência. 
`min_max` calcula os dois e devolve uma tupla de dois valores.

```python
min_max([2, 4, 1, 3])
```

Podemos atribuir os resultados a variáveis assim:

```python
low, high = min_max([2, 4, 1, 3])
low, high
```

## Empacotamento de argumentos

As funções podem receber um número variável de argumentos. 
Um nome de parâmetro que começa com o operador `*` **empacota** os argumentos em uma tupla.
Por exemplo, a função a seguir recebe qualquer número de argumentos e calcula a média aritmética: a soma deles dividida pelo número de argumentos. 

```python
def mean(*args):
    return sum(args) / len(args)
```

O parâmetro pode ter o nome que você quiser, mas `args` é o convencional.
Podemos chamar a função assim:

```python
mean(1, 2, 3)
```

Se você tem uma sequência de valores e quer passá-los a uma função como vários argumentos, pode usar o operador `*` para **desempacotar** a tupla.
Por exemplo, `divmod` recebe exatamente dois argumentos: se você passar uma tupla como parâmetro, recebe um erro.

```python
%%expect TypeError
t = (7, 3)
divmod(t)
```

Mesmo que a tupla contenha dois elementos, ela conta como um único argumento.
Mas se você desempacota a tupla, ela é tratada como dois argumentos.

```python
divmod(*t)
```

Empacotar e desempacotar pode ser útil se você quer adaptar o comportamento de uma função existente.
Por exemplo, esta função recebe qualquer número de argumentos, remove o menor e o maior, e calcula a média dos demais.

```python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
```

Primeiro ela usa `min_max` para encontrar o menor e o maior elemento.
Depois converte `args` em uma lista para poder usar o método `remove`.
Por fim, desempacota a lista para que os elementos sejam passados a `mean` como argumentos separados, e não como uma única lista.

Aqui está um exemplo que mostra o efeito.

```python
mean(1, 2, 3, 10)
```

```python
trimmed_mean(1, 2, 3, 10)
```

Esse tipo de média "aparada" é usada em alguns esportes com julgamento subjetivo, como saltos ornamentais e ginástica, para reduzir o efeito de um juiz cuja nota se desvia das demais. 

## Zip

As tuplas são úteis para percorrer os elementos de duas sequências e realizar operações nos elementos correspondentes.
Por exemplo, suponha que duas equipes joguem uma série de sete partidas e registremos as pontuações em duas listas, uma para cada equipe.

```python
scores1 = [1, 2, 4, 5, 1, 5, 2]
scores2 = [5, 5, 2, 2, 5, 2, 3]
```

Vamos ver quantos jogos cada equipe venceu.
Usaremos `zip`, que é uma função nativa que recebe duas ou mais sequências e devolve um **objeto zip**, assim chamado porque emparelha os elementos das sequências como os dentes de um zíper.

```python
zip(scores1, scores2)
```

Podemos usar o objeto zip para percorrer os valores das sequências em pares.

```python
for pair in zip(scores1, scores2):
     print(pair)
```

A cada passagem pelo laço, `pair` recebe uma tupla de pontuações.
Assim, podemos atribuir as pontuações a variáveis e contar as vitórias da primeira equipe, assim:

```python
wins = 0
for team1, team2 in zip(scores1, scores2):
    if team1 > team2:
        wins += 1
        
wins
```

Infelizmente, a primeira equipe venceu só três jogos e perdeu a série.

Se você tem duas listas e quer uma lista de pares, pode usar `zip` e `list`.

```python
t = list(zip(scores1, scores2))
t
```

O resultado é uma lista de tuplas, então podemos obter o resultado do último jogo assim:

```python
t[-1]
```

Se você tem uma lista de chaves e uma lista de valores, pode usar `zip` e `dict` para montar um dicionário.
Por exemplo, eis como podemos criar um dicionário que mapeia cada letra para sua posição no alfabeto.

```python
letters = 'abcdefghijklmnopqrstuvwxyz'
numbers = range(len(letters))
letter_map = dict(zip(letters, numbers))
```

Agora podemos procurar uma letra e obter seu índice no alfabeto.

```python
letter_map['a'], letter_map['z']
```

Nesse mapeamento, o índice de `'a'` é `0` e o índice de `'z'` é `25`.

Se você precisa percorrer os elementos de uma sequência e seus índices, pode usar a função nativa `enumerate`.

```python
enumerate('abc')
```

O resultado é um **objeto enumerate** que percorre uma sequência de pares, em que cada par contém um índice (começando em 0) e um elemento da sequência dada.

```python
for index, element in enumerate('abc'):
    print(index, element)
```

## Comparação e ordenação

Os operadores relacionais funcionam com tuplas e outras sequências.
Por exemplo, se você usa o operador `<` com tuplas, ele começa comparando o primeiro elemento de cada sequência.
Se forem iguais, segue para o próximo par de elementos, e assim por diante, até encontrar um par que diferem. 

```python
(0, 1, 2) < (0, 3, 4)
```

Os elementos seguintes não são considerados, mesmo que sejam bem grandes.

```python
(0, 1, 2000000) < (0, 3, 4)
```

Esse jeito de comparar tuplas é útil para ordenar uma lista de tuplas, ou encontrar o mínimo ou o máximo.
Como exemplo, vamos encontrar a letra mais comum em uma palavra.
No capítulo anterior, escrevemos `value_counts`, que recebe uma string e devolve um dicionário que mapeia cada letra para o número de vezes em que ela aparece.

```python
def value_counts(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

Aqui está o resultado para a string `'banana'`.

```python
counter = value_counts('banana')
counter
```

Com só três itens, vemos facilmente que a letra mais frequente é `'a'`, que aparece três vezes.
Mas se houvesse mais itens, seria útil ordená-los automaticamente.

Podemos obter os itens de `counter` assim.

```python
items = counter.items()
items
```

O resultado é um objeto `dict_items` que se comporta como uma lista de tuplas, então podemos ordená-lo assim.

```python
sorted(items)
```

O comportamento padrão é usar o primeiro elemento de cada tupla para ordenar a lista, e usar o segundo elemento para desempatar.

No entanto, para encontrar os itens com as maiores contagens, queremos usar o segundo elemento para ordenar a lista.
Podemos fazer isso escrevendo uma função que recebe uma tupla e devolve o segundo elemento.

```python
def second_element(t):
    return t[1]
```

Em seguida, podemos passar essa função a `sorted` como um argumento opcional chamado `key`, que indica que essa função deve ser usada para calcular a **chave de ordenação** de cada item.

```python
sorted_items = sorted(items, key=second_element)
sorted_items
```

A chave de ordenação determina a ordem dos itens na lista.
A letra com a menor contagem aparece primeiro, e a letra com a maior contagem aparece por último.
Assim, podemos encontrar a letra mais comum assim.

```python
sorted_items[-1]
```

Se só queremos o máximo, não precisamos ordenar a lista.
Podemos usar `max`, que também recebe `key` como argumento opcional.

```python
max(items, key=second_element)
```

Para encontrar a letra com a menor contagem, poderíamos usar `min` do mesmo jeito.

## Invertendo um dicionário

Suponha que você queira inverter um dicionário para poder procurar um valor e obter a chave correspondente.
Por exemplo, se você tem um contador de palavras que mapeia cada palavra para o número de vezes em que ela aparece, poderia criar um dicionário que mapeia de inteiros para as palavras que aparecem aquele número de vezes.

Mas há um problema: as chaves de um dicionário precisam ser únicas, mas os valores não. Por exemplo, em um contador de palavras, pode haver muitas palavras com a mesma contagem.

Então, uma forma de inverter um dicionário é criar um dicionário novo em que os valores são listas de chaves do original.
Como exemplo, vamos contar as letras em `parrot`.

```python
d =  value_counts('parrot')
d
```

Se invertermos este dicionário, o resultado deve ser `{1: ['p', 'a', 'o', 't'], 2: ['r']}`, o que indica que as letras que aparecem uma vez são `'p'`, `'a'`, `'o'` e `'t'`, e a letra que aparece duas vezes é `'r'`.

A função a seguir recebe um dicionário e devolve o inverso como um dicionário novo.

```python
def invert_dict(d):
    new = {}
    for key, value in d.items():
        if value not in new:
            new[value] = [key]
        else:
            new[value].append(key)
    return new
```

A instrução `for` percorre as chaves e os valores em `d`.
Se o valor ainda não está no dicionário novo, ele é adicionado e associado a uma lista que contém um único elemento.
Caso contrário, é acrescentado à lista existente.

Podemos testá-la assim:

```python
invert_dict(d)
```

E obtemos o resultado esperado.

Este é o primeiro exemplo em que vimos valores de dicionário que são listas.

## Glossário

**empacotar:**
Reunir vários argumentos em uma tupla.

**desempacotar:**
Tratar uma tupla (ou outra sequência) como vários argumentos.

**objeto zip:**
O resultado de chamar a função nativa `zip`; pode ser usado para percorrer uma sequência de tuplas.

**objeto enumerate:**
O resultado de chamar a função nativa `enumerate`; pode ser usado para percorrer uma sequência de tuplas.

**chave de ordenação:**
Um valor, ou uma função que calcula um valor, usado para ordenar os elementos de uma coleção.

**estrutura de dados:**
Uma coleção de valores, organizada para realizar certas operações com eficiência.
