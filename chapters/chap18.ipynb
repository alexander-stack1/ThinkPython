## Conjuntos

Python oferece uma classe chamada `set` que representa uma coleção de elementos únicos.
Para criar um conjunto vazio, podemos usar o objeto de classe como uma função.

```python
s1 = set()
s1
```

Podemos usar o método `add` para adicionar elementos.

```python
s1.add('a')
s1.add('b')
s1
```

Ou podemos passar qualquer tipo de sequência a `set`.

```python
s2 = set('acd')
s2
```

Um elemento só pode aparecer uma vez em um `set`.
Se você adicionar um elemento que já está lá, isso não tem efeito.

```python
s1.add('a')
s1
```

Ou, se criar um conjunto com uma sequência que contém duplicatas, o resultado contém só elementos únicos.

```python
set('banana')
```

Alguns exercícios deste livro podem ser feitos de forma concisa e eficiente com conjuntos.
Por exemplo, aqui está a solução de um exercício do capítulo 11 que usa um dicionário para verificar se há elementos duplicados em uma sequência.

```python
def has_duplicates(t):
    d = {}
    for x in t:
        d[x] = True
    return len(d) < len(t)
```

Esta versão adiciona os elementos de `t` como chaves em um dicionário e depois verifica se há menos chaves do que elementos.
Usando conjuntos, podemos escrever a mesma função assim.

```python
def has_duplicates(t):
    s = set(t)
    return len(s) < len(t)
```

```python
has_duplicates('abba')
```

Um elemento só pode aparecer uma vez em um conjunto, então, se um elemento de `t` aparecer mais de uma vez, o conjunto será menor que `t`.
Se não houver duplicatas, o conjunto terá o mesmo tamanho que `t`.

Objetos `set` oferecem métodos que realizam operações de conjunto.
Por exemplo, `union` calcula a união de dois conjuntos, que é um conjunto novo contendo todos os elementos que aparecem em qualquer um dos dois.

```python
s1.union(s2)
```

Alguns operadores aritméticos funcionam com conjuntos.
Por exemplo, o operador `-` faz a subtração de conjuntos: o resultado é um conjunto novo que contém todos os elementos do primeiro conjunto que *não* estão no segundo.

```python
s1 - s2
```

No [capítulo 12](section_dictionary_subtraction) usamos dicionários para encontrar as palavras que aparecem em um documento, mas não em uma lista de palavras.
Usamos a função a seguir, que recebe dois dicionários e devolve um dicionário novo contendo só as chaves do primeiro que não aparecem no segundo.

```python
def subtract(d1, d2):
    res = {}
    for key in d1:
        if key not in d2:
            res[key] = d1[key]
    return res
```

Com conjuntos, não precisamos escrever essa função nós mesmos.
Se `word_counter` for um dicionário que contém as palavras únicas do documento e `word_list` for uma lista de palavras válidas, podemos calcular a diferença de conjuntos assim.

```python
# esta célula cria um exemplo pequeno para podermos executar a célula
# seguinte sem carregar os dados reais

word_counter = {'word': 1}
word_list = ['word']
```

```python
set(word_counter) - set(word_list)
```

O resultado é um conjunto que contém as palavras do documento que não aparecem na lista de palavras.

Os operadores relacionais funcionam com conjuntos.
Por exemplo, `<=` verifica se um conjunto é subconjunto de outro, inclusive a possibilidade de serem iguais.

```python
set('ab') <= set('abc')
```

Com esses operadores, podemos usar conjuntos para fazer alguns dos exercícios do capítulo 7.
Por exemplo, aqui está uma versão de `uses_only` que usa um laço.

```python
def uses_only(word, available):
    for letter in word: 
        if letter not in available:
            return False
    return True
```

`uses_only` verifica se todas as letras de `word` estão em `available`.
Com conjuntos, podemos reescrevê-la assim.

```python
def uses_only(word, available):
    return set(word) <= set(available)
```

Se as letras de `word` forem um subconjunto das letras de `available`, isso significa que `word` usa só letras de `available`.

## Counters

Um `Counter` é como um conjunto, exceto que, se um elemento aparecer mais de uma vez, o `Counter` registra quantas vezes ele aparece.
Se você conhece a ideia matemática de "multiconjunto", um `Counter` é um
jeito natural de representar um multiconjunto.

A classe `Counter` está definida em um módulo chamado `collections`, então você precisa importá-la.
Depois pode usar o objeto de classe como uma função e passar como argumento uma string, uma lista ou qualquer outro tipo de sequência.

```python
from collections import Counter

counter = Counter('banana')
counter
```

```python
from collections import Counter

t = (1, 1, 1, 2, 2, 3)
counter = Counter(t)
counter
```

Um objeto `Counter` é como um dicionário que mapeia cada chave para o número de vezes em que ela aparece.
Como nos dicionários, as chaves precisam ser hashable.

Diferente dos dicionários, objetos `Counter` não levantam uma exceção se você acessar um
elemento que não aparece.
Em vez disso, devolvem `0`.

```python
counter['d']
```

Podemos usar objetos `Counter` para resolver um dos exercícios do capítulo 10, que pede uma função que recebe duas palavras e verifica se são anagramas: isto é, se as letras de uma podem ser rearranjadas para soletrar a outra.

Aqui está uma solução usando objetos `Counter`.

```python
def is_anagram(word1, word2):
    return Counter(word1) == Counter(word2)
```

Se duas palavras são anagramas, elas contêm as mesmas letras com as mesmas contagens, então os objetos `Counter` delas são equivalentes.

`Counter` oferece um método chamado `most_common` que devolve uma lista de pares valor-frequência, ordenada da mais comum para a menos comum.

```python
counter.most_common()
```

Eles também oferecem métodos e operadores para realizar operações semelhantes às de conjuntos, inclusive adição, subtração, união e interseção.
Por exemplo, o operador `+` combina dois objetos `Counter` e cria um `Counter` novo que contém as chaves dos dois e as somas das contagens.

Podemos testá-lo criando um `Counter` com as letras de `'bans'` e somando-o às letras de `'banana'`.

```python
counter2 = Counter('bans')
counter + counter2
```

Você vai ter a chance de explorar outras operações de `Counter` nos exercícios do fim deste capítulo.

## defaultdict

O módulo `collections` também oferece `defaultdict`, que é como um dicionário, exceto que, se você acessar uma chave que não existe, ele gera um valor novo automaticamente.

Quando você cria um `defaultdict`, fornece uma função usada para criar valores novos.
Uma função que cria objetos às vezes se chama **fábrica**.
As funções nativas que criam listas, conjuntos e outros tipos podem ser usadas como fábricas.

Por exemplo, aqui está um `defaultdict` que cria uma `list` nova quando necessário.

```python
from collections import defaultdict

d = defaultdict(list)
d
```

Observe que o argumento é `list`, que é um objeto de classe, não `list()`, que é uma chamada de função que cria uma lista nova.
A função fábrica não é chamada a menos que acessemos uma chave que não existe.

```python
t = d['new key']
t
```

A lista nova, que estamos chamando de `t`, também é adicionada ao dicionário.
Então, se modificarmos `t`, a mudança aparece em `d`:

```python
t.append('new value')
d['new key']
```

Se você estiver fazendo um dicionário de listas, muitas vezes pode escrever código mais
simples usando `defaultdict`.

Em um dos exercícios do [capítulo 11](chapter_tuples), fiz um dicionário que mapeia de uma string de letras ordenadas para a lista de palavras que podem ser soletradas com essas letras.
Por exemplo, a string `'opst'` mapeia para a lista `['opts', 'post', 'pots', 'spot', 'stop', 'tops']`.
Aqui está o código original.

```python
def all_anagrams(filename):
    d = {}
    for line in open(filename):
        word = line.strip().lower()
        t = signature(word)
        if t not in d:
            d[t] = [word]
        else:
            d[t].append(word)
    return d
```

E aqui está uma versão mais simples usando um `defaultdict`.

```python
def all_anagrams(filename):
    d = defaultdict(list)
    for line in open(filename):
        word = line.strip().lower()
        t = signature(word)
        d[t].append(word)
    return d
```

## Expressões condicionais

Instruções condicionais muitas vezes são usadas para escolher um de dois valores, assim:

```python
import math
x = 5
```

```python
if x > 0:
    y = math.log(x)
else:
    y = float('nan')
```

```python
y
```

Esta instrução verifica se `x` é positivo. Se for, calcula o logaritmo.
Se não for, `math.log` levantaria um ValueError.
Para evitar interromper o programa, geramos um `NaN`, que é um valor especial de ponto flutuante que representa "Not a Number" (não é um número).

Podemos escrever esta instrução de forma mais concisa usando uma **expressão condicional**.

```python
y = math.log(x) if x > 0 else float('nan')
```

```python
y
```

Você quase pode ler esta linha como se fosse inglês: "`y` recebe log-`x` se `x` for maior que 0; caso contrário, recebe `NaN`".

Funções recursivas às vezes podem ser escritas de forma concisa usando expressões condicionais.
Por exemplo, aqui está uma versão de `factorial` com uma *instrução* condicional.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

E aqui está uma versão com uma *expressão* condicional.

```python
def factorial(n):
    return 1 if n == 0 else n * factorial(n-1)
```

Outro uso de expressões condicionais é tratar argumentos opcionais.
Por exemplo, aqui está uma definição de classe com um método `__init__` que usa uma instrução condicional para verificar um parâmetro com valor padrão.

```python
class Kangaroo:
    def __init__(self, name, contents=None):
        self.name = name
        if contents is None:
            contents = []
        self.contents = contents
```

Aqui está uma versão que usa uma expressão condicional.

```python
def __init__(self, name, contents=None):
    self.name = name
    self.contents = [] if contents is None else contents 
```

Em geral, você pode substituir uma instrução condicional por uma expressão condicional se os dois ramos contiverem uma única expressão e nenhuma instrução.

## Compreensões de lista

Nos capítulos anteriores, vimos alguns exemplos em que começamos com uma lista vazia e adicionamos elementos, um de cada vez, usando o método `append`.
Por exemplo, suponha que tenhamos uma string com o título de um filme e queiramos colocar todas as palavras em maiúsculas.

```python
title = 'monty python and the holy grail'
```

Podemos dividi-la em uma lista de strings, percorrer as strings, capitalizá-las e acrescentá-las a uma lista.

Podemos fazer a mesma coisa de forma mais concisa usando uma **compreensão de lista**:

```python
t = [word.capitalize() for word in title.split()]

' '.join(t)
```

Os operadores de colchete indicam que estamos construindo uma lista nova.
A expressão dentro dos colchetes especifica os elementos da lista, e a cláusula `for` indica por qual sequência estamos iterando.

A sintaxe de uma compreensão de lista pode parecer estranha, porque a variável do laço, `word` neste exemplo, aparece na expressão antes de chegarmos à definição dela.
Mas você se acostuma.

Como outro exemplo, no [capítulo 9](section_word_list) usamos este laço para ler palavras de um arquivo e acrescentá-las a uma lista.

```python
download('https://raw.githubusercontent.com/AllenDowney/ThinkPython2/master/code/words.txt');
```

```python
word_list = []

for line in open('words.txt'):
    word = line.strip()
    word_list.append(word)
```

```python
len(word_list)
```

Aqui está como podemos escrever isso como uma compreensão de lista.

```python
word_list = [line.strip() for line in open('words.txt')]
```

```python
len(word_list)
```

Uma compreensão de lista também pode ter uma cláusula `if` que determina quais elementos entram na lista.
Por exemplo, aqui está um laço `for` que usamos no [capítulo 10](section_palindrome_list) para fazer uma lista só das palavras de `word_list` que são palíndromos.

```python
def is_palindrome(word):
    return list(reversed(word)) == list(word)
```

```python
palindromes = []

for word in word_list:
    if is_palindrome(word):
        palindromes.append(word)
```

```python
palindromes[:10]
```

Aqui está como podemos fazer a mesma coisa com uma compreensão de lista.

```python
palindromes = [word for word in word_list if is_palindrome(word)]
```

```python
palindromes[:10]
```

Quando uma compreensão de lista é usada como argumento de uma função, muitas vezes podemos omitir os colchetes.
Por exemplo, suponha que queiramos somar $1 / 2^n$ para valores de $n$ de 0 a 9.
Podemos usar uma compreensão de lista assim.

```python
sum([1/2**n for n in range(10)])
```

Ou podemos deixar de fora os colchetes, assim.

```python
sum(1/2**n for n in range(10))
```

Neste exemplo, o argumento é tecnicamente uma **expressão geradora**, não uma compreensão de lista, e ela nunca chega a criar uma lista.
Mas, tirando isso, o comportamento é o mesmo.

Compreensões de lista e expressões geradoras são concisas e fáceis de ler, pelo menos para expressões simples.
E em geral são mais rápidas que os laços `for` equivalentes, às vezes bem mais rápidas.
Então, se você estiver bravo comigo por não tê-las mencionado antes, eu entendo.

Mas, em minha defesa, compreensões de lista são mais difíceis de depurar porque você não pode colocar uma instrução `print` dentro do laço.
Sugiro usá-las só se o cálculo for simples o bastante para você provavelmente acertar de primeira.
Ou considere escrever e depurar um laço `for` e depois convertê-lo em uma compreensão de lista.

## `any` e `all`

Python oferece uma função nativa, `any`, que recebe uma sequência de valores booleanos e devolve `True` se algum dos valores for `True`.

```python
any([False, False, True])
```

`any` muitas vezes é usada com expressões geradoras.

```python
any(letter == 't' for letter in 'monty')
```

Esse exemplo não é muito útil porque faz a mesma coisa que o operador `in`.
Mas poderíamos usar `any` para escrever soluções concisas de alguns exercícios do [capítulo 7](chapter_search). Por exemplo, podemos escrever `uses_none` assim.

```python
def uses_none(word, forbidden):
    """Checks whether a word avoids forbidden letters."""
    return not any(letter in forbidden for letter in word)
```

```python
uses_none('banana', 'xyz')
```

```python
uses_none('apple', 'efg')
```

Esta função percorre as letras de `word` e verifica se alguma delas está em `forbidden`.
Usar `any` com uma expressão geradora é eficiente porque ela para imediatamente se encontrar um valor `True`, então não precisa percorrer a sequência inteira.

Python oferece outra função nativa, `all`, que devolve `True` se todos os elementos da sequência forem `True`.
Podemos usá-la para escrever uma versão concisa de `uses_all`.

```python
def uses_all(word, required):
    """Check whether a word uses all required letters."""
    return all(letter in word for letter in required)
```

```python
uses_all('banana', 'ban')
```

```python
uses_all('apple', 'api')
```

Expressões que usam `any` e `all` podem ser concisas, eficientes e fáceis de ler.

## Tuplas nomeadas

O módulo `collections` oferece uma função chamada `namedtuple` que pode ser usada para criar classes simples.
Por exemplo, o objeto `Point` no [capítulo 16](section_create_point) tem só dois atributos, `x` e `y`.
Aqui está como o definimos.

```python
class Point:
    """Represents a point in 2-D space."""
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f'({self.x}, {self.y})'
```

Isso é muito código para transmitir pouca informação.
`namedtuple` oferece um jeito mais conciso de definir classes assim.

```python
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
```

O primeiro argumento é o nome da classe que você quer criar. O
segundo é uma lista dos atributos que os objetos `Point` devem ter.
O resultado é um objeto de classe, por isso é atribuído a uma variável com nome capitalizado.

Uma classe criada com `namedtuple` oferece um método `__init__` que atribui valores aos atributos e um `__str__` que exibe o objeto de forma legível.
Assim podemos criar e exibir um objeto `Point` assim.

```python
p = Point(1, 2)
p
```

`Point` também oferece um método `__eq__` que verifica se dois objetos `Point` são equivalentes: isto é, se os atributos são os mesmos.

```python
p == Point(1, 2)
```

Você pode acessar os elementos de uma tupla nomeada pelo nome ou pelo índice.

```python
p.x, p.y
```

```python
p[0], p[1]
```

Você também pode tratar uma tupla nomeada como uma tupla, como nesta atribuição.

```python
x, y = p
x, y
```

Mas objetos `namedtuple` são imutáveis.
Depois que os atributos são inicializados, não podem ser alterados.

```python
%%expect TypeError

p[0] = 3
```

```python
%%expect AttributeError

p.x = 3
```

`namedtuple` oferece um jeito rápido de definir classes simples.
O inconveniente é que classes simples nem sempre continuam simples.
Você pode decidir depois que quer adicionar métodos a uma tupla nomeada.
Nesse caso, pode definir uma classe nova que herda da tupla nomeada.

```python
class Pointier(Point):
    """This class inherits from Point"""
```

## Empacotando argumentos nomeados

No [capítulo 11](section_argument_pack), escrevemos uma função que empacota os argumentos em uma tupla.

```python
def mean(*args):
    return sum(args) / len(args)
```

Você pode chamar esta função com qualquer número de argumentos.

```python
mean(1, 2, 3)
```

Mas o operador `*` não empacota argumentos nomeados.
Então chamar esta função com um argumento nomeado causa um erro.

```python
%%expect TypeError

mean(1, 2, start=3)
```

Para empacotar argumentos nomeados, podemos usar o operador `**`:

```python
def mean(*args, **kwargs):
    print(kwargs)
    return sum(args) / len(args)
```

O parâmetro de empacotamento de argumentos nomeados pode ter qualquer nome, mas `kwargs` é uma escolha comum.
O resultado é um dicionário que mapeia de palavras-chave para valores.

```python
mean(1, 2, start=3)
```

Neste exemplo, o valor de `kwargs` é impresso, mas, fora isso, não tem efeito.

Mas o operador `**` também pode ser usado em uma lista de argumentos para desempacotar um dicionário.
Por exemplo, aqui está uma versão de `mean` que empacota quaisquer argumentos nomeados que receber e depois os desempacota como argumentos nomeados para `sum`.

```python
def mean(*args, **kwargs):
    return sum(args, **kwargs) / len(args)
```

Agora, se chamarmos `mean` com `start` como argumento nomeado, ele é repassado a `sum`, que o usa como ponto de partida da soma.
No exemplo a seguir, `start=3` adiciona `3` à soma antes de calcular a média, então a soma é `6` e o resultado é `3`.

```python
mean(1, 2, start=3)
```

Como outro exemplo, se tivermos um dicionário com as chaves `x` e `y`, podemos usá-lo com o operador de desempacotamento para criar um objeto `Point`.

```python
d = dict(x=1, y=2)
Point(**d)
```

Sem o operador de desempacotamento, `d` é tratado como um único argumento posicional, então é atribuído a `x`, e obtemos um `TypeError` porque não há um segundo argumento para atribuir a `y`.

```python
%%expect TypeError

d = dict(x=1, y=2)
Point(d)
```

Quando você trabalha com funções que têm um grande número de argumentos nomeados, muitas vezes é útil criar e passar dicionários que especificam opções usadas com frequência.

```python
def pack_and_print(**kwargs):
    print(kwargs)
    
pack_and_print(a=1, b=2)
```

## Depuração

Nos capítulos anteriores, usamos `doctest` para testar funções.
Por exemplo, aqui está uma função chamada `add` que recebe dois números e devolve a soma.
Ela inclui um doctest que verifica se `2 + 2` é `4`.

```python
def add(a, b):
    '''Add two numbers.
    
    >>> add(2, 2)
    4
    '''
    return a + b
```

Esta função recebe um objeto função e executa os doctests dele.

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```

Então podemos testar `add` assim.

```python
run_doctests(add)
```

Não há saída, o que significa que todos os testes passaram.

Python oferece outra ferramenta para executar testes automatizados, chamada `unittest`.
É um pouco mais complicada de usar, mas aqui vai um exemplo.

```python
from unittest import TestCase

class TestExample(TestCase):

    def test_add(self):
        result = add(2, 2)
        self.assertEqual(result, 4)
```

Primeiro importamos `TestCase`, que é uma classe do módulo `unittest`.
Para usá-la, temos de definir uma classe nova que herda de `TestCase` e oferece pelo menos um método de teste.
O nome do método de teste deve começar com `test` e deve indicar qual função ele testa.

Neste exemplo, `test_add` testa a função `add` chamando-a, guardando o resultado e invocando `assertEqual`, que é herdado de `TestCase`.
`assertEqual` recebe dois argumentos e verifica se são iguais.

Para executar este método de teste, temos de executar uma função de `unittest` chamada `main` e fornecer vários argumentos nomeados.
A função a seguir mostra os detalhes. Se tiver curiosidade, você pode pedir a um assistente virtual que explique como ela funciona.

```python
import unittest

def run_unittest():
    unittest.main(argv=[''], verbosity=0, exit=False)
```

`run_unittest` não recebe `TestExample` como argumento: em vez disso, procura classes que herdam de `TestCase`.
Depois procura métodos que começam com `test` e os executa.
Esse processo se chama **descoberta de testes**.

Aqui está o que acontece quando chamamos `run_unittest`.

```python
run_unittest()
```

`unittest.main` relata o número de testes que executou e os resultados.
Neste caso, `OK` indica que os testes passaram.

Para ver o que acontece quando um teste falha, vamos adicionar um método de teste incorreto a `TestExample`.

```python
%%add_method_to TestExample

    def test_add_broken(self):
        result = add(2, 2)
        self.assertEqual(result, 100)
```

Aqui está o que acontece quando executamos os testes.

```python
run_unittest()
```

O relatório inclui o método de teste que falhou e uma mensagem de erro mostrando onde.
O resumo indica que dois testes rodaram e um falhou.

## Glossário

**fábrica:**
Função usada para criar objetos, muitas vezes passada como parâmetro a uma função.

**expressão condicional:**
Expressão que usa uma condicional para escolher um de dois valores.

Jeito conciso de percorrer uma sequência e criar uma lista.

Semelhante a uma compreensão de lista, exceto que não cria uma lista.

Processo usado para encontrar e executar testes.

A seguinte é uma função que calcula o coeficiente binomial
de forma recursiva.

```python
def binomial_coeff(n, k):
    """Compute the binomial coefficient "n choose k".

    n: number of trials
    k: number of successes

    returns: int
    """
    if k == 0:
        return 1
    
    if n == 0:
        return 0

    return binomial_coeff(n-1, k) + binomial_coeff(n-1, k-1)
```
