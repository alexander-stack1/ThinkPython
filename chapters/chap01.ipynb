## Operadores aritméticos

Um **operador aritmético** é um símbolo que representa uma computação aritmética. Por exemplo, o sinal de mais, `+`, realiza a adição.

```python
30 + 12
```

O sinal de menos, `-`, é o operador que realiza a subtração.

```python
43 - 1
```

O asterisco, `*`, realiza a multiplicação.

```python
6 * 7
```

E a barra, `/`, realiza a divisão:

```python
84 / 2
```

Observe que o resultado da divisão é `42.0`, e não `42`. Isso acontece porque há dois tipos de números em Python:

* **inteiros**, que representam números sem parte fracionária ou decimal, e

* **números de ponto flutuante**, que representam inteiros e números com ponto decimal.

Se você soma, subtrai ou multiplica dois inteiros, o resultado é um inteiro.
Mas se divide dois inteiros, o resultado é um número de ponto flutuante.
O Python oferece outro operador, `//`, que realiza a **divisão inteira**.
O resultado da divisão inteira é sempre um inteiro.

```python
84 // 2
```

A divisão inteira também é chamada de "divisão pelo piso" porque sempre arredonda para baixo (em direção ao "piso").

```python
85 // 2
```

Por fim, o operador `**` realiza a exponenciação, isto é, eleva um
número a uma potência:

```python
7 ** 2
```

Em algumas outras linguagens, o circunflexo, `^`, é usado para exponenciação, mas em Python
ele é um operador bit a bit chamado XOR.
Se você não conhece operadores bit a bit, o resultado pode surpreender:

```python
7 ^ 2
```

Não vou tratar de operadores bit a bit neste livro, mas você pode ler sobre
eles em <http://wiki.python.org/moin/BitwiseOperators>.

## Expressões

Uma coleção de operadores e números se chama **expressão**.
Uma expressão pode conter qualquer quantidade de operadores e números.
Por exemplo, esta expressão contém dois operadores.

```python
6 + 6 ** 2
```

Observe que a exponenciação acontece antes da adição.
O Python segue a ordem das operações que você talvez tenha aprendido na aula de matemática: a exponenciação acontece antes da multiplicação e da divisão, que acontecem antes da adição e da subtração.

No exemplo a seguir, a multiplicação acontece antes da adição.

```python
12 + 5 * 6
```

Se você quiser que a adição aconteça primeiro, pode usar parênteses.

```python
(12 + 5) * 6
```

Toda expressão tem um **valor**.
Por exemplo, a expressão `6 * 7` tem o valor `42`.

## Funções aritméticas

Além dos operadores aritméticos, o Python oferece algumas **funções** que trabalham com números.
Por exemplo, a função `round` recebe um número de ponto flutuante e o arredonda para o inteiro mais próximo.

```python
round(42.4)
```

```python
round(42.6)
```

A função `abs` calcula o valor absoluto de um número.
Para um número positivo, o valor absoluto é o próprio número.

```python
abs(42)
```

Para um número negativo, o valor absoluto é positivo.

```python
abs(-42)
```

Quando usamos uma função assim, dizemos que estamos **chamando** a função.
Uma expressão que chama uma função é uma **chamada de função**.

Quando você chama uma função, os parênteses são obrigatórios.
Se você os omitir, recebe uma mensagem de erro.

## Strings

Além de números, o Python também pode representar sequências de letras, chamadas **strings** porque as letras são enfileiradas como contas num colar.
Para escrever uma string, podemos colocar uma sequência de letras entre aspas retas.

```python
'Hello'
```

Também é válido usar aspas duplas.

```python
"world"
```

As aspas duplas facilitam escrever uma string que contém um apóstrofo, que é o mesmo símbolo da aspa reta.

```python
"it's a small "
```

Strings também podem conter espaços, pontuação e dígitos.

```python
'Well, '
```

O operador `+` funciona com strings; ele junta duas strings numa string só, o que se chama **concatenação**.

```python
'Well, ' + "it's a small " + 'world.'
```

O operador `*` também funciona com strings; ele faz várias cópias de uma string e as concatena.

```python
'Spam, ' * 4
```

Os outros operadores aritméticos não funcionam com strings.

O Python oferece uma função chamada `len` que calcula o comprimento de uma string.

```python
len('Spam')
```

Observe que `len` conta as letras entre as aspas, mas não as aspas.

Quando criar uma string, use aspas retas.
A crase, também chamada de *backtick*, causa um erro de sintaxe.

```python
%%expect SyntaxError

`Hello`
```

Aspas tipográficas, também chamadas de aspas curvas, também são inválidas.

```python
%%expect SyntaxError

‘Hello’
```

## Valores e tipos

Até agora vimos três espécies de valores:

* `2` é um inteiro,

* `42.0` é um número de ponto flutuante, e

* `'Hello'` é uma string.

Uma espécie de valor se chama **tipo**.
Todo valor tem um tipo, ou às vezes dizemos que ele "pertence a" um tipo.

O Python oferece uma função chamada `type` que informa o tipo de qualquer valor.
O tipo de um inteiro é `int`.

```python
type(2)
```

O tipo de um número de ponto flutuante é `float`.

```python
type(42.0)
```

E o tipo de uma string é `str`.

```python
type('Hello, World!')
```

Os tipos `int`, `float` e `str` podem ser usados como funções.
Por exemplo, `int` pode receber um número de ponto flutuante e convertê-lo em um inteiro (sempre arredondando para baixo).

```python
int(42.9)
```

E `float` pode converter um inteiro em um valor de ponto flutuante.

```python
float(42)
```

Agora, eis algo que pode confundir.
O que você obtém se colocar uma sequência de dígitos entre aspas?

```python
'126'
```

Parece um número, mas na verdade é uma string.

```python
type('126')
```

Se você tentar usá-la como um número, pode receber um erro.

```python
%%expect TypeError

'126' / 3
```

Este exemplo gera um `TypeError`, o que significa que os valores da expressão, chamados **operandos**, têm o tipo errado.
A mensagem de erro indica que o operador `/` não admite os tipos desses valores, que são `str` e `int`.

Se você tem uma string que contém dígitos, pode usar `int` para convertê-la em um inteiro.

```python
int('126') / 3
```

Se você tem uma string que contém dígitos e um ponto decimal, pode usar `float` para convertê-la em um número de ponto flutuante.

```python
float('12.6')
```

Ao escrever um inteiro grande, você pode ter vontade de usar vírgulas
entre grupos de dígitos, como em `1,000,000`.
Essa é uma expressão válida em Python, mas o resultado não é um inteiro.

```python
1,000,000
```

O Python interpreta `1,000,000` como uma sequência de inteiros separados por vírgula.
Vamos aprender mais sobre esse tipo de sequência mais adiante.

Você pode usar sublinhados para deixar números grandes mais fáceis de ler.

```python
1_000_000
```

## Linguagens formais e naturais

**Linguagens naturais** são as línguas que as pessoas falam, como inglês, espanhol e francês. Elas não foram projetadas por pessoas; evoluíram naturalmente.

**Linguagens formais** são linguagens projetadas por pessoas para aplicações específicas.
Por exemplo, a notação que matemáticos usam é uma linguagem formal especialmente boa para denotar relações entre números e símbolos.
Da mesma forma, linguagens de programação são linguagens formais projetadas para expressar computações.

Embora linguagens formais e naturais tenham alguns traços em
comum, há diferenças importantes:

* Ambiguidade: as linguagens naturais são cheias de ambiguidade, com a qual as pessoas lidam
    usando pistas contextuais e outras informações. As linguagens formais são
    projetadas para ser quase ou completamente sem ambiguidade, o que significa que
    qualquer programa tem exatamente um significado, independentemente do contexto.

* Redundância: para compensar a ambiguidade e reduzir mal-entendidos,
    as linguagens naturais usam redundância. Por isso, costumam ser
    verbosas. As linguagens formais são menos redundantes e mais concisas.

* Literalidade: as linguagens naturais são cheias de expressões idiomáticas e metáforas. As linguagens formais significam exatamente o que dizem.

Como todos crescemos falando linguagens naturais, às vezes é difícil se adaptar às linguagens formais.
As linguagens formais são mais densas que as naturais, então levam mais tempo para serem lidas.
Além disso, a estrutura importa, então nem sempre o melhor é ler de cima para baixo, da esquerda para a direita.
Por fim, os detalhes importam. Pequenos erros de ortografia e
pontuação, que passam em linguagens naturais, podem fazer
uma grande diferença numa linguagem formal.

## Depuração

Quem programa comete erros. Por razões caprichosas, erros de programação se chamam **bugs** e o processo de rastreá-los se chama **depuração**.

Programar, e especialmente depurar, às vezes desperta emoções fortes. Se você está lutando com um bug difícil, pode se sentir com raiva, triste ou envergonhado.

Preparar-se para essas reações pode ajudar a lidar com elas. Uma abordagem é pensar no computador como um funcionário com certas forças, como velocidade e precisão, e fraquezas particulares, como falta de empatia e incapacidade de ver o quadro geral.

Seu trabalho é ser um bom gestor: encontrar formas de aproveitar as forças e mitigar as fraquezas. E encontrar formas de usar suas emoções para se engajar no problema, sem deixar que as reações atrapalhem a capacidade de trabalhar com eficácia.

Aprender a depurar pode ser frustrante, mas é uma habilidade valiosa, útil em muitas atividades além da programação. No fim de cada capítulo há uma seção como esta, com minhas sugestões de depuração. Espero que ajudem!

## Glossário

**operador aritmético:**
Um símbolo, como `+` e `*`, que denota uma operação aritmética como adição ou multiplicação.

**inteiro:**
Um tipo que representa números sem parte fracionária ou decimal.

**ponto flutuante:**
Um tipo que representa inteiros e números com parte decimal.

**divisão inteira:**
Um operador, `//`, que divide dois números e arredonda para baixo até um inteiro.

**expressão:**
Uma combinação de variáveis, valores e operadores.

**valor:**
Um inteiro, um número de ponto flutuante ou uma string, ou um dos outros tipos de valor que veremos depois.

**função:**
Uma sequência nomeada de instruções que realiza alguma operação útil.
Funções podem ou não receber argumentos e podem ou não produzir um resultado.

**chamada de função:**
Uma expressão, ou parte de uma expressão, que executa uma função.
Consiste no nome da função seguido de uma lista de argumentos entre parênteses.

**erro de sintaxe:**
Um erro num programa que o torna impossível de analisar e, portanto, impossível de executar.

**string:**
 Um tipo que representa sequências de caracteres.

**concatenação:**
Juntar duas strings ponta a ponta.

**tipo:**
Uma categoria de valores.
Os tipos que vimos até agora são inteiros (tipo `int`), números de ponto flutuante (tipo `float`) e strings (tipo `str`).

**operando:**
Um dos valores sobre os quais um operador opera.

**linguagem natural:**
Qualquer uma das línguas que as pessoas falam e que evoluíram naturalmente.

**linguagem formal:**
Qualquer uma das linguagens que as pessoas projetaram para fins específicos, como representar ideias matemáticas ou programas de computador.
Todas as linguagens de programação são linguagens formais.

**bug:**
Um erro num programa.

**depuração:**
O processo de encontrar e corrigir erros.
