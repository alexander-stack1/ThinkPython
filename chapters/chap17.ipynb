# Herança

O recurso de linguagem mais associado à programação orientada a objetos é a **herança**.
Herança é a capacidade de definir uma classe nova que é uma versão modificada de uma classe existente.
Neste capítulo, demonstro herança usando classes que representam cartas de baralho, baralhos e mãos de pôquer.
Se você não joga pôquer, não se preocupe: vou dizer o que você precisa saber.

## Representando cartas

Há 52 cartas em um baralho padrão: cada uma pertence a um de quatro naipes e a um de treze valores.
Os naipes são Espadas (Spades), Copas (Hearts), Ouros (Diamonds) e Paus (Clubs).
Os valores são Ás (Ace), 2, 3, 4, 5, 6, 7, 8, 9, 10, Valete (Jack), Dama (Queen) e Rei (King).
Conforme o jogo, um Ás pode valer mais que o Rei ou menos que o 2.

Se quisermos definir um objeto novo para representar uma carta, é óbvio quais devem ser os atributos: `rank` e `suit`.
É menos óbvio de que tipo esses atributos devem ser.
Uma possibilidade é usar strings como `'Spade'` para naipes e `'Queen'` para valores.
Um problema dessa implementação é que não seria fácil comparar cartas para ver qual tem valor ou naipe maior.

Uma alternativa é usar inteiros para **codificar** os valores e os naipes.
Neste contexto, "codificar" significa que vamos definir um mapeamento entre números e naipes, ou entre números e valores.
Esse tipo de codificação não pretende ser um segredo (isso seria "criptografia").

Por exemplo, esta tabela mostra os naipes e os códigos inteiros correspondentes:


| Naipe | Código |
| --- | --- |
|  Spades     |   3  |
|  Hearts     |   2  |
|  Diamonds   |   1  |
|  Clubs      |   0  |

Com essa codificação, podemos comparar naipes comparando os códigos.

Para codificar os valores, vamos usar o inteiro `2` para representar o valor `2`, `3` para representar `3`, e assim por diante até `10`.
A tabela a seguir mostra os códigos das figuras.

  
| Valor | Código |
| --- | --- |
|  Jack     |   11  |
|  Queen   |   12  |
|  King      |   13  |

E podemos usar `1` ou `14` para representar um Ás, conforme queiramos considerá-lo menor ou maior que os outros valores.

Para representar essas codificações, vamos usar duas listas de strings, uma com os nomes dos naipes e outra com os nomes dos valores.

Aqui está a definição de uma classe que representa uma carta, com essas listas de strings como **variáveis de classe**, que são variáveis definidas dentro de uma definição de classe, mas não dentro de um método.

```python
class Card:
    """Represents a standard playing card."""

    suit_names = ['Clubs', 'Diamonds', 'Hearts', 'Spades']
    rank_names = [None, 'Ace', '2', '3', '4', '5', '6', '7', 
                  '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']
```

O primeiro elemento de `rank_names` é `None` porque não existe carta de valor zero. Incluindo `None` como marcador de posição, obtemos uma lista com a propriedade conveniente de que o índice `2` mapeia para a string `'2'`, e assim por diante.

Variáveis de classe estão associadas à classe, e não a uma instância da classe, então podemos acessá-las assim.

```python
Card.suit_names
```

Podemos usar `suit_names` para consultar um naipe e obter a string correspondente.

```python
Card.suit_names[0]
```

E `rank_names` para consultar um valor.

```python
Card.rank_names[11]
```

## Comparando cartas

Suponha que criemos um segundo objeto `Card` com o mesmo naipe e o mesmo valor.

```python
queen2 = Card(1, 12)
print(queen2)
```

Se usarmos o operador `==` para compará-los, ele verifica se `queen` e `queen2` se referem ao mesmo objeto.

```python
queen == queen2
```

Não se referem, então devolve `False`.
Podemos mudar esse comportamento definindo o método especial `__eq__`.

```python
%%add_method_to Card

    def __eq__(self, other):
        return self.suit == other.suit and self.rank == other.rank
```

`__eq__` recebe dois objetos `Card` como parâmetros e devolve `True` se tiverem o mesmo naipe e o mesmo valor, mesmo que não sejam o mesmo objeto.
Em outras palavras, verifica se são equivalentes, mesmo que não sejam idênticos.

Quando usamos o operador `==` com objetos `Card`, Python chama o método `__eq__`.

```python
queen == queen2
```

Como segundo teste, vamos criar uma carta com o mesmo naipe e um valor diferente.

```python
six = Card(1, 6)
print(six)
```

Podemos confirmar que `queen` e `six` não são equivalentes.

```python
queen == six
```

Se usarmos o operador `!=`, Python invoca um método especial chamado `__ne__`, se ele existir.
Caso contrário, invoca `__eq__` e inverte o resultado: então, se `__eq__` devolver `True`, o resultado do operador `!=` é `False`.

```python
queen != queen2
```

```python
queen != six
```

Agora suponha que queiramos comparar duas cartas para ver qual é maior.
Se usarmos um dos operadores relacionais, obtemos um `TypeError`.

```python
%%expect TypeError

queen < queen2
```

Para mudar o comportamento do operador `<`, podemos definir um método especial chamado `__lt__`, abreviação de "less than" (menor que).
Para este exemplo, vamos supor que o naipe é mais importante que o valor: assim, todas as Espadas vencem todas as Copas, que vencem todos os Ouros, e assim por diante.
Se duas cartas tiverem o mesmo naipe, vence a de valor maior.

Para implementar essa lógica, vamos usar o método a seguir, que devolve uma tupla contendo o naipe e o valor da carta, nessa ordem.

```python
%%add_method_to Card

    def to_tuple(self):
        return (self.suit, self.rank)
```

Podemos usar esse método para escrever `__lt__`.

```python
%%add_method_to Card

    def __lt__(self, other):
        return self.to_tuple() < other.to_tuple()
```

A comparação de tuplas compara os primeiros elementos de cada tupla, que representam os naipes.
Se forem iguais, compara os segundos elementos, que representam os valores.

Agora, se usarmos o operador `<`, ele invoca o método `__lt__`.

```python
six < queen
```

Se usarmos o operador `>`, ele invoca um método especial chamado `__gt__`, se ele existir.
Caso contrário, invoca `__lt__` com os argumentos na ordem inversa.

```python
queen < queen2
```

```python
queen > queen2
```

Por fim, se usarmos o operador `<=`, ele invoca um método especial chamado `__le__`.

```python
%%add_method_to Card

    def __le__(self, other):
        return self.to_tuple() <= other.to_tuple()
```

Assim podemos verificar se uma carta é menor ou igual a outra.

```python
queen <= queen2
```

```python
queen <= six
```

Se usarmos o operador `>=`, ele usa `__ge__` se existir. Caso contrário, invoca `__le__` com os argumentos na ordem inversa.

```python
queen >= six
```

Como os definimos, esses métodos são completos no sentido de que podemos comparar quaisquer dois objetos `Card`, e consistentes no sentido de que os resultados de operadores diferentes não se contradizem.
Com essas duas propriedades, podemos dizer que os objetos `Card` são **totalmente ordenados**.
E isso significa, como vamos ver em breve, que eles podem ser ordenados.

## Baralhos

Agora que temos objetos que representam cartas, vamos definir objetos que representam baralhos.
A seguir está uma definição de classe para `Deck` com
um método `__init__` que recebe uma lista de objetos `Card` como parâmetro e a atribui a um atributo chamado `cards`.

```python
class Deck:

    def __init__(self, cards):
        self.cards = cards
```

Para criar uma lista com as 52 cartas de um baralho padrão, vamos usar o método estático a seguir.

```python
%%add_method_to Deck

    def make_cards():
        cards = []
        for suit in range(4):
            for rank in range(2, 15):
                card = Card(suit, rank)
                cards.append(card)
        return cards
```

Em `make_cards`, o laço externo enumera os naipes de `0` a `3`.
O laço interno enumera os valores de `2` a `14`, em que `14` representa um Ás que vale mais que um Rei.
Cada iteração cria um `Card` novo com o naipe e o valor atuais e o acrescenta a `cards`.

Aqui está como fazemos uma lista de cartas e um objeto `Deck` que a contém.

```python
cards = Deck.make_cards()
deck = Deck(cards)
len(deck.cards)
```

Ele contém 52 cartas, como pretendido.

## Imprimindo o baralho

Aqui está um método `__str__` para `Deck`.

```python
%%add_method_to Deck

    def __str__(self):
        res = []
        for card in self.cards:
            res.append(str(card))
        return '\n'.join(res)
```

Esse método demonstra um jeito eficiente de acumular uma string grande: montar uma lista de strings e depois usar o método de string `join`.

Vamos testar esse método com um baralho que contém só duas cartas.

```python
small_deck = Deck([queen, six])
```

Se chamarmos `str`, ele invoca `__str__`.

```python
str(small_deck)
```

Quando o Jupyter exibe uma string, mostra a forma "representacional" da string, que representa uma quebra de linha com a sequência `\n`.

Porém, se imprimirmos o resultado, o Jupyter mostra a forma "imprimível" da string, que imprime a quebra de linha como espaço em branco.

```python
print(small_deck)
```

Assim as cartas aparecem em linhas separadas.

## Adicionar, remover, embaralhar e ordenar

Para distribuir cartas, gostaríamos de um método que removesse uma carta do baralho
e a devolvesse. O método de lista `pop` oferece um jeito conveniente de fazer
isso.

```python
%%add_method_to Deck

    def take_card(self):
        return self.cards.pop()
```

Aqui está como o usamos.

```python
card = deck.take_card()
print(card)
```

Podemos confirmar que restam `51` cartas no baralho.

```python
len(deck.cards)
```

Para adicionar uma carta, podemos usar o método de lista `append`.

```python
%%add_method_to Deck

    def put_card(self, card):
        self.cards.append(card)
```

Como exemplo, podemos devolver a carta que acabamos de retirar.

```python
deck.put_card(card)
len(deck.cards)
```

Para embaralhar o baralho, podemos usar a função `shuffle` do módulo `random`:

```python
import random
```

```python
# Esta célula inicializa o gerador de números aleatórios para que
# sempre obtenhamos os mesmos resultados.

random.seed(3)
```

```python
%%add_method_to Deck
            
    def shuffle(self):
        random.shuffle(self.cards)
```

Se embaralharmos o baralho e imprimirmos as primeiras cartas, vemos que elas não estão em nenhuma ordem aparente.

```python
deck.shuffle()
for card in deck.cards[:4]:
    print(card)
```

Para ordenar as cartas, podemos usar o método de lista `sort`, que ordena os elementos "no lugar": isto é, modifica a lista em vez de criar uma lista nova.

```python
%%add_method_to Deck
            
    def sort(self):
        self.cards.sort()
```

Quando invocamos `sort`, ele usa o método `__lt__` para comparar cartas.

```python
deck.sort()
```

Se imprimirmos as primeiras cartas, podemos confirmar que estão em ordem crescente.

```python
for card in deck.cards[:4]:
    print(card)
```

Neste exemplo, `Deck.sort` não faz nada além de invocar `list.sort`.
Passar a responsabilidade assim se chama **delegação**.

## Pais e filhos

Herança é a capacidade de definir uma classe nova que é uma versão modificada de uma classe existente.
Como exemplo, digamos que queiramos uma classe para representar uma "mão", isto é, as cartas que um jogador segura.

* Uma mão é parecida com um baralho: as duas são feitas de uma coleção de cartas, e as duas exigem operações como adicionar e remover cartas.

* Uma mão também é diferente de um baralho: há operações que queremos para mãos e que não fazem sentido para um baralho. Por exemplo, no pôquer podemos comparar duas mãos para ver qual vence. No bridge, podemos calcular a pontuação de uma mão para fazer um lance.

Essa relação entre classes, em que uma é uma versão especializada da outra, se presta à herança.

Para definir uma classe nova com base em uma classe existente, colocamos o nome da classe existente entre parênteses.

```python
class Hand(Deck):
    """Represents a hand of playing cards."""
```

Essa definição indica que `Hand` herda de `Deck`, o que significa que objetos `Hand` podem acessar métodos definidos em `Deck`, como `take_card` e `put_card`.

`Hand` também herda `__init__` de `Deck`, mas, se definirmos `__init__` na classe `Hand`, ele sobrescreve o da classe `Deck`.

```python
%%add_method_to Hand

    def __init__(self, label=''):
        self.label = label
        self.cards = []
```

Esta versão de `__init__` recebe uma string opcional como parâmetro e sempre começa com uma lista vazia de cartas.
Quando criamos um `Hand`, Python invoca este método, não o de `Deck`, o que podemos confirmar verificando que o resultado tem um atributo `label`.

```python
hand = Hand('player 1')
hand.label
```

Para distribuir uma carta, podemos usar `take_card` para remover uma carta de um `Deck` e `put_card` para adicionar a carta a um `Hand`.

```python
deck = Deck(cards)
card = deck.take_card()
hand.put_card(card)
print(hand)
```

Vamos encapsular este código em um método de `Deck` chamado `move_cards`.

```python
%%add_method_to Deck

    def move_cards(self, other, num):
        for i in range(num):
            card = self.take_card()
            other.put_card(card)
```

Este método é polimórfico: isto é, funciona com mais de um tipo. `self` e `other` podem ser um `Hand` ou um `Deck`.
Assim podemos usar este método para distribuir uma carta de um `Deck` para um `Hand`, de um `Hand` para outro, ou de um `Hand` de volta para um `Deck`.

Quando uma classe nova herda de uma existente, a existente se chama **pai** e a nova se chama **filha**. Em geral:

* Instâncias da classe filha devem ter todos os atributos da classe pai, mas podem ter atributos adicionais.

* A classe filha deve ter todos os métodos da classe pai, mas pode ter métodos adicionais.

* Se uma classe filha sobrescrever um método da classe pai, o método novo deve receber os mesmos parâmetros e devolver um resultado compatível.

Esse conjunto de regras se chama "princípio da substituição de Liskov", em homenagem à cientista da computação Barbara Liskov.

Se você seguir essas regras, qualquer função ou método projetado para funcionar com uma instância de uma classe pai, como um `Deck`, também vai funcionar com instâncias de uma classe filha, como `Hand`.
Se violar essas regras, o código desaba como um castelo de cartas (desculpe).

## Especialização

Vamos criar uma classe chamada `BridgeHand` que representa uma mão de bridge, um jogo de cartas muito jogado.
Vamos herdar de `Hand` e adicionar um método novo chamado `high_card_point_count` que avalia uma mão pelo método de "pontos de honra", que soma pontos pelas cartas altas da mão.

Aqui está uma definição de classe que contém, como variável de classe, um dicionário que mapeia nomes de cartas aos respectivos valores em pontos.

```python
class BridgeHand(Hand):
    """Represents a bridge hand."""

    hcp_dict = {
        'Ace': 4,
        'King': 3,
        'Queen': 2,
        'Jack': 1,
    }
```

Dado o valor de uma carta, como `12`, podemos usar `Card.rank_names` para obter a representação em string do valor e depois usar `hcp_dict` para obter a pontuação.

```python
rank = 12
rank_name = Card.rank_names[rank]
score = BridgeHand.hcp_dict.get(rank_name, 0)
rank_name, score
```

O método a seguir percorre as cartas de um `BridgeHand` e soma as pontuações.

```python
%%add_method_to BridgeHand

    def high_card_point_count(self):
        count = 0
        for card in self.cards:
            rank_name = Card.rank_names[card.rank]
            count += BridgeHand.hcp_dict.get(rank_name, 0)
        return count
```

```python
# Esta célula cria um Deck novo e 
# inicializa o gerador de números aleatórios

cards = Deck.make_cards()
deck = Deck(cards)
random.seed(3)
```

Para testá-lo, vamos distribuir uma mão com cinco cartas. Uma mão de bridge em geral tem treze, mas é mais fácil testar código com exemplos pequenos.

```python
hand = BridgeHand('player 2')

deck.shuffle()
deck.move_cards(hand, 5)
print(hand)
```

E aqui está a pontuação total do Rei e da Dama.

```python
hand.high_card_point_count()
```

`BridgeHand` herda as variáveis e os métodos de `Hand` e adiciona uma variável de classe e um método específicos do bridge.
Esse modo de usar herança se chama **especialização**, porque define uma classe nova especializada para um uso particular, como jogar bridge.

## Depuração

Herança é um recurso útil.
Alguns programas que seriam repetitivos sem herança podem ser escritos de forma mais concisa com ela.
Além disso, herança pode facilitar o reúso de código, pois você pode personalizar o comportamento de uma classe pai sem precisar modificá-la.
Em alguns casos, a estrutura de herança reflete a estrutura natural do problema, o que torna o projeto mais fácil de entender.

Por outro lado, herança pode tornar os programas difíceis de ler.
Quando um método é invocado, às vezes não fica claro onde encontrar a definição: o código relevante pode estar espalhado por vários módulos.

Sempre que você não tiver certeza sobre o fluxo de execução do programa, a solução mais simples é adicionar instruções `print` no início dos métodos relevantes.
Se `Deck.shuffle` imprimir uma mensagem como `Running Deck.shuffle`, então, enquanto o programa roda, ele traça o fluxo de execução.

Como alternativa, você pode usar a função a seguir, que recebe um objeto e o nome de um método (como string) e devolve a classe que fornece a definição do método.

```python
def find_defining_class(obj, method_name):
    """Find the class where the given method is defined."""
    for typ in type(obj).mro():
        if method_name in vars(typ):
            return typ
    return f'Method {method_name} not found.'
```

`find_defining_class` usa o método `mro` para obter a lista de objetos de classe (tipos) que serão pesquisados em busca de métodos.
"MRO" significa "method resolution order" (ordem de resolução de métodos), que é a sequência de classes que Python pesquisa para "resolver" o nome de um método: isto é, para encontrar o objeto função a que o nome se refere.

Como exemplo, vamos instanciar um `BridgeHand` e depois encontrar a classe que define `shuffle`.

```python
hand = BridgeHand('player 3')
find_defining_class(hand, 'shuffle')
```

O método `shuffle` do objeto `BridgeHand` é o de `Deck`.

## Glossário

**herança:**
A capacidade de definir uma classe nova que é uma versão modificada de uma classe definida antes.

**codificar:**
Representar um conjunto de valores usando outro conjunto de valores, construindo um mapeamento entre eles.

**variável de classe:**
Variável definida dentro de uma definição de classe, mas não dentro de nenhum método.

**totalmente ordenado:**
Um conjunto de objetos é totalmente ordenado se pudermos comparar quaisquer dois elementos e os resultados forem consistentes.

**delegação:**
Quando um método passa a responsabilidade a outro método para fazer a maior parte do trabalho, ou todo ele.

**classe pai:**
Classe da qual se herda.

**classe filha:**
Classe que herda de outra classe.

**especialização:**
Modo de usar herança para criar uma classe nova que é uma versão especializada de uma classe existente.

Os próximos exercícios pedem que você escreva funções que classifiquem mãos de pôquer.
Se você não conhece pôquer, vou explicar o que precisa saber.
Vamos usar a classe a seguir para representar mãos de pôquer.

```python
class PokerHand(Hand):
    """Represents a poker hand."""

    def get_suit_counts(self):
        counter = {}
        for card in self.cards:
            key = card.suit
            counter[key] = counter.get(key, 0) + 1
        return counter
    
    def get_rank_counts(self):
        counter = {}
        for card in self.cards:
            key = card.rank
            counter[key] = counter.get(key, 0) + 1
        return counter    
```

`PokerHand` oferece dois métodos que vão ajudar nos exercícios.

* `get_suit_counts` percorre as cartas do `PokerHand`, conta o número de cartas de cada naipe e devolve um dicionário que mapeia cada código de naipe para o número de vezes em que aparece.

* `get_rank_counts` faz a mesma coisa com os valores das cartas, devolvendo um dicionário que mapeia cada código de valor para o número de vezes em que aparece.

Um Ás pode vir antes do dois ou depois do Rei, então `Ace`, `2`, `3`, `4`, `5` é uma sequência, e `10`, `Jack`, `Queen`, `King`, `Ace` também é.
Mas uma sequência não pode "dar a volta", então `King`, `Ace`, `2`, `3`, `4` não é uma sequência.

Você pode usar o esboço a seguir para começar.
Ele inclui algumas linhas de código que contam o número de Ases, representados com o código `1` ou `14`, e guardam o total nas duas posições do contador.

```python
%%add_method_to PokerHand

    def has_straight(self, n=5):
        """Checks whether this hand has a straight with at least `n` cards."""
        counter = self.get_rank_counts()
        aces = counter.get(1, 0) + counter.get(14, 0)
        counter[1] = aces
        counter[14] = aces
        
        return False
```

### Exercício

Este exercício é um conto de advertência sobre um erro comum que pode ser difícil de depurar.
Considere a definição de classe a seguir.

```python
class Kangaroo:
    """A Kangaroo is a marsupial."""
    
    def __init__(self, name, contents=[]):
        """Initialize the pouch contents.

        name: string
        contents: initial pouch contents.
        """
        self.name = name
        self.contents = contents

    def __str__(self):
        """Return a string representaion of this Kangaroo.
        """
        t = [ self.name + ' has pouch contents:' ]
        for obj in self.contents:
            s = '    ' + object.__str__(obj)
            t.append(s)
        return '\n'.join(t)

    def put_in_pouch(self, item):
        """Adds a new item to the pouch contents.

        item: object to be added
        """
        self.contents.append(item)
```

`__init__` recebe dois parâmetros: `name` é obrigatório, mas `contents` é opcional. Se não for fornecido, o valor padrão é uma lista vazia.

`__str__` devolve uma representação em string do objeto que inclui o nome e o conteúdo da bolsa.

`put_in_pouch` recebe qualquer objeto e o acrescenta a `contents`.

Agora vamos ver como essa classe funciona.
Vamos criar dois objetos `Kangaroo` com os nomes `'Kanga'` e `'Roo'`.

```python
kanga = Kangaroo('Kanga')
roo = Kangaroo('Roo')
```

Na bolsa da Kanga vamos colocar duas strings e o Roo.

```python
kanga.put_in_pouch('wallet')
kanga.put_in_pouch('car keys')
kanga.put_in_pouch(roo)
```

Se imprimirmos `kanga`, parece que tudo funcionou.

```python
print(kanga)
```

Mas o que acontece se imprimirmos `roo`?

```python
print(roo)
```

A bolsa do Roo contém o mesmo conteúdo que a da Kanga, inclusive uma referência a `roo`!
