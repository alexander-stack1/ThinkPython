## Divisão inteira e módulo

Lembre-se de que o operador de divisão inteira, `//`, divide dois números e arredonda
para baixo até um inteiro.
Por exemplo, suponha que a duração de um filme é 105 minutos.
Você pode querer saber quanto isso é em horas.
A divisão convencional devolve um número de ponto flutuante:

```python
minutes = 105
minutes / 60
```

Mas normalmente não escrevemos horas com pontos decimais.
A divisão inteira devolve o número inteiro de horas, arredondando para baixo:

```python
minutes = 105
hours = minutes // 60
hours
```

Para obter o resto, você poderia subtrair uma hora em minutos:

```python
remainder = minutes - hours * 60
remainder
```

Ou poderia usar o **operador módulo**, `%`, que divide dois números e devolve o resto.

```python
remainder = minutes % 60
remainder
```

O operador módulo é mais útil do que parece.
Por exemplo, ele pode verificar se um número é divisível por outro: se `x % y` for zero, então `x` é divisível por `y`.

Também pode extrair o dígito ou os dígitos mais à direita de um número.
Por exemplo, `x % 10` produz o dígito mais à direita de `x` (na base 10).
Da mesma forma, `x % 100` produz os dois últimos dígitos.

```python
x = 123
x % 10
```

```python
x % 100
```

Por fim, o operador módulo pode fazer "aritmética de relógio".
Por exemplo, se um evento começa às 11h e dura três horas, podemos usar o operador módulo para descobrir a que horas termina.

```python
start = 11
duration = 3
end = (start + duration) % 12
end
```

O evento terminaria às 14h.

## Operadores lógicos

Para combinar valores booleanos em expressões, podemos usar **operadores lógicos**.
Os mais comuns são `and`, ` or` e `not`.
O sentido desses operadores é semelhante ao sentido em inglês.
Por exemplo, o valor da expressão a seguir é `True` só se `x` for maior que `0` *e* menor que `10`.

```python
x > 0 and x < 10
```

A expressão a seguir é `True` se *uma ou as duas* condições forem verdadeiras, isto é, se o número for divisível por 2 *ou* por 3:

```python
x % 2 == 0 or x % 3 == 0
```

Por fim, o operador `not` nega uma expressão booleana, então a expressão a seguir é `True` se `x > y` for `False`.

```python
not x > y
```

Em sentido estrito, os operandos de um operador lógico deveriam ser expressões booleanas, mas o Python não é muito rigoroso.
Qualquer número diferente de zero é interpretado como `True`:

```python
42 and True
```

Essa flexibilidade pode ser útil, mas há sutilezas que podem confundir.

## Instruções if

Para escrever programas úteis, quase sempre precisamos da capacidade de
verificar condições e mudar o comportamento do programa de acordo com elas.
**Instruções condicionais** nos dão essa capacidade. A forma mais simples é
a instrução `if`:

```python
if x > 0:
    print('x is positive')
```

`if` é uma palavra-chave de Python.
Instruções `if` têm a mesma estrutura que definições de função: um
cabeçalho seguido de uma instrução recuada ou de uma sequência de instruções chamada **bloco**.

A expressão booleana depois de `if` se chama **condição**.
Se for verdadeira, as instruções do bloco recuado executam. Se não for, não executam.

Não há limite para o número de instruções que podem aparecer no bloco, mas precisa haver pelo menos uma.
Às vezes é útil ter um bloco que não faz nada, em geral como marcador de lugar para código que você ainda não escreveu.
Nesse caso, você pode usar a instrução `pass`, que não faz nada.

```python
if x < 0:
    pass          # TODO: é preciso tratar valores negativos!
```

A palavra `TODO` num comentário é um lembrete convencional de que há algo que você precisa fazer depois.

## A cláusula `else`

Uma instrução `if` pode ter uma segunda parte, chamada cláusula `else`.
A sintaxe fica assim:

```python
if x % 2 == 0:
    print('x is even')
else:
    print('x is odd')
```

Se a condição for verdadeira, a primeira instrução recuada executa; caso contrário, a segunda instrução recuada executa.

Neste exemplo, se `x` for par, o resto da divisão de `x` por `2` é `0`, então a condição é verdadeira e o programa exibe `x is even`.
Se `x` for ímpar, o resto é `1`, então a condição
é falsa e o programa exibe `x is odd`.

Como a condição precisa ser verdadeira ou falsa, exatamente uma das alternativas vai executar.
As alternativas se chamam **ramos**.

## Condicionais encadeados

Às vezes há mais de duas possibilidades e precisamos de mais de dois ramos.
Um jeito de expressar uma computação assim é um **condicional encadeado**, que inclui uma cláusula `elif`.

```python
if x < y:
    print('x is less than y')
elif x > y:
    print('x is greater than y')
else:
    print('x and y are equal')
```

`elif` é uma abreviação de "else if".
Não há limite para o número de cláusulas `elif`.
Se houver uma cláusula `else`, ela precisa ficar no fim, mas não precisa haver
uma.

Cada condição é verificada em ordem.
Se a primeira for falsa, a próxima é verificada, e assim por diante.
Se uma delas for verdadeira, o ramo correspondente executa e a instrução `if` termina.
Mesmo que mais de uma condição seja verdadeira, só o primeiro ramo verdadeiro executa.

## Condicionais aninhados

Um condicional também pode ser aninhado dentro de outro.
Poderíamos ter escrito o exemplo da seção anterior assim:

```python
if x == y:
    print('x and y are equal')
else:
    if x < y:
        print('x is less than y')
    else:
        print('x is greater than y')
```

A instrução `if` externa contém dois ramos.
O primeiro ramo contém uma instrução simples. O segundo ramo contém outra instrução `if`, que tem dois ramos próprios.
Esses dois ramos são ambos instruções simples, embora também pudessem ter sido instruções condicionais.

Embora o recuo das instruções deixe a estrutura aparente, **condicionais aninhados** podem ser difíceis de ler.
Sugiro que você os evite quando puder.

Operadores lógicos muitas vezes oferecem um jeito de simplificar instruções condicionais aninhadas.
Aqui está um exemplo com um condicional aninhado.

```python
if 0 < x:
    if x < 10:
        print('x is a positive single-digit number.')
```

A instrução `print` só executa se passarmos pelos dois condicionais, então obtemos o mesmo efeito com o operador `and`.

```python
if 0 < x and x < 10:
    print('x is a positive single-digit number.')
```

Para este tipo de condição, o Python oferece uma opção mais concisa:

```python
if 0 < x < 10:
    print('x is a positive single-digit number.')
```

## Recursão

É válido uma função chamar a si mesma.
Pode não ser óbvio por que isso é uma boa ideia, mas acaba sendo uma das coisas mais mágicas que um programa pode fazer.
Aqui está um exemplo.

```python
def countdown(n):
    if n <= 0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n-1)
```

Se `n` for 0 ou negativo, `countdown` exibe a palavra "Blastoff!". Caso contrário, ela
exibe `n` e depois chama a si mesma, passando `n-1` como argumento.

Veja o que acontece quando chamamos esta função com o argumento `3`.

```python
countdown(3)
```

A execução de `countdown` começa com `n=3` e, como `n` é maior
que `0`, ela exibe `3` e depois chama a si mesma\...

> A execução de `countdown` começa com `n=2` e, como `n` é
> maior que `0`, ela exibe `2` e depois chama a si mesma\...
>
> > A execução de `countdown` começa com `n=1` e, como `n` é
> > maior que `0`, ela exibe `1` e depois chama a si mesma\...
> >
> > > A execução de `countdown` começa com `n=0` e, como `n`
> > > não é maior que `0`, ela exibe "Blastoff!" e retorna.
> >
> > O `countdown` que recebeu `n=1` retorna.
>
> O `countdown` que recebeu `n=2` retorna.

O `countdown` que recebeu `n=3` retorna.

Uma função que chama a si mesma é **recursiva**.
Como outro exemplo, podemos escrever uma função que imprime uma string `n` vezes.

```python
def print_n_times(string, n):
    if n > 0:
        print(string)
        print_n_times(string, n-1)
```

Se `n` for positivo, `print_n_times` exibe o valor de `string` e depois chama a si mesma, passando `string` e `n-1` como argumentos.

Se `n` for `0` ou negativo, a condição é falsa e `print_n_times` não faz nada.

Veja como funciona.

```python
print_n_times('Spam ', 4)
```

Para exemplos simples como este, provavelmente é mais fácil usar um laço
`for`. Mas vamos ver depois exemplos que são difíceis de escrever com um laço
`for` e fáceis de escrever com recursão, então é bom começar cedo.

Aqui está um diagrama de pilha que mostra os quadros criados quando chamamos `countdown` com `n = 3`.

Ela não faz uma chamada recursiva, então não há mais quadros.

## Recursão infinita

Se uma recursão nunca chega a um caso-base, ela continua fazendo chamadas
recursivas para sempre, e o programa nunca termina. Isso se chama
**recursão infinita**, e em geral não é uma boa ideia.
Aqui está uma função mínima com uma recursão infinita.

```python
def recurse():
    recurse()
```

Toda vez que `recurse` é chamada, ela chama a si mesma, o que cria outro quadro.
Em Python, há um limite para o número de quadros que podem estar na pilha ao mesmo tempo.
Se um programa ultrapassa o limite, causa um erro de tempo de execução.

```python
%xmode Context
```

```python
%%expect RecursionError

recurse()
```

O traceback indica que havia quase 3000 quadros na pilha quando o erro ocorreu.

Se você encontrar uma recursão infinita por acidente, revise a função para confirmar que há um caso-base que não faz uma chamada recursiva. E se houver um caso-base, verifique se você tem garantia de chegar a ele.

## Glossário

**recursão:**
O processo de chamar a função que está em execução no momento.

**operador módulo:**
Um operador, `%`, que atua sobre inteiros e devolve o resto da divisão de um número por outro.

**expressão booleana:**
Uma expressão cujo valor é `True` ou `False`.

**operador relacional:**
Um dos operadores que comparam seus operandos: `==`, `!=`, `>`, `<`, `>=` e `<=`.

**operador lógico:**
Um dos operadores que combinam expressões booleanas, inclusive `and`, `or` e `not`.

**instrução condicional:**
Uma instrução que controla o fluxo de execução conforme alguma condição.

**condição:**
A expressão booleana numa instrução condicional que determina qual ramo executa.

**bloco:**
Uma ou mais instruções recuadas para indicar que fazem parte de outra instrução.

**ramo:**
Uma das sequências alternativas de instruções numa instrução condicional.

**condicional encadeado:**
Uma instrução condicional com uma série de ramos alternativos.

**condicional aninhado:**
Uma instrução condicional que aparece num dos ramos de outra instrução condicional.

**recursiva:**
Uma função que chama a si mesma é recursiva.

**caso-base:**
Um ramo condicional numa função recursiva que não faz uma chamada recursiva.

**recursão infinita:**
Uma recursão que não tem caso-base, ou nunca chega a ele.
Com o tempo, uma recursão infinita causa um erro de tempo de execução.

**quebra de linha:**
Um caractere que cria uma quebra de linha entre duas partes de uma string.
