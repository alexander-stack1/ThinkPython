## O que é um notebook?

Um notebook Jupyter é feito de células, e cada célula contém texto ou código.
Esta célula contém texto.

A célula seguinte contém código.

```python
print('Hello')
```

Clique na célula anterior para selecioná-la.
Você deve ver à esquerda um botão com um triângulo dentro de um círculo, o ícone de "Reproduzir".
Se pressionar esse botão, o Jupyter executa o código da célula e mostra o resultado.

Na primeira vez em que você executa código em um notebook, pode levar alguns segundos para iniciar.
E, se for um notebook que você não escreveu, pode aparecer um aviso.
Se estiver executando um notebook de uma fonte em que você confia, e espero que isso me inclua, pode pressionar "Run Anyway".

Em vez de clicar no botão "Reproduzir", você também pode executar o código de uma célula segurando `Shift` e pressionando `Enter`.

Se estiver executando este notebook no Colab, deve ver no canto superior esquerdo os botões "+ Code" e "+ Text". O primeiro adiciona uma célula de código e o segundo, uma célula de texto.
Se quiser experimentar, selecione esta célula clicando nela e depois pressione "+ Text".
Uma célula nova deve aparecer abaixo desta.

Adicione algum texto à célula.
Você pode usar os botões para formatá-lo, ou marcar o texto com [Markdown](https://www.markdownguide.org/basic-syntax/).
Quando terminar, segure `Shift` e pressione `Enter`: o texto que você digitou será formatado e o cursor irá para a célula seguinte.

O Jupyter está sempre em um de dois modos:

* No **modo de comando**, você executa operações que afetam células, como adicionar e remover células inteiras.

* No **modo de edição**, você edita o conteúdo de uma célula.

Nas células de texto, fica óbvio em que modo você está.
No modo de edição, a célula se divide na vertical, com o texto que você edita à esquerda e o texto formatado à direita.
E você vê ferramentas de edição de texto na parte de cima.
No modo de comando, você vê só o texto formatado.

Nas células de código, a diferença é mais sutil, mas se houver um cursor na célula você está no modo de edição.

Para ir do modo de edição ao modo de comando, pressione `ESC`.
Para ir do modo de comando ao modo de edição, pressione `Enter`.

Quando terminar de trabalhar em um notebook, você pode fechar a janela, mas as alterações desaparecerão.
Se fizer alguma alteração que queira guardar, abra o menu File no canto superior esquerdo.
Você verá várias formas de salvar o notebook.

* Se tiver uma conta Google, pode salvar o notebook no Drive.

* Se tiver uma conta no GitHub, pode salvá-lo no GitHub.

* Ou, se quiser salvar o notebook no computador, escolha "Download" e depois "Download .ipynb". O sufixo ".ipynb" indica que é um arquivo de notebook, e não ".py", que indica um arquivo só com código Python.

## Código de *Pense em Python*

No começo de cada notebook, você verá uma célula com código assim:

```python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/thinkpython.py')

import thinkpython
```

Você não precisa saber como esse código funciona, mas quando chegar ao fim do livro a maior parte fará sentido.
Como você já imagina, ele baixa um arquivo, especificamente o `thinkpython.py`, que contém código Python feito para este livro.
A última linha "importa" esse código, o que significa que podemos usá-lo no notebook.

Em outros capítulos, você verá código que baixa `diagram.py`, usado para gerar os diagramas do livro, e `jupyturtle.py`, usado em vários capítulos para criar gráficos de tartaruga.

Em alguns pontos você verá uma célula como esta, que começa com `%%expect`.

```python
%%expect SyntaxError

abs 42
```

`%%expect` não faz parte de Python: é um "comando mágico" do Jupyter que indica que esperamos que a célula produza um erro.
Quando você vir esse comando, o erro é deliberado, em geral para alertar sobre uma armadilha comum.
