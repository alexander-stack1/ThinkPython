## Laços e strings

No Capítulo 3 vimos um laço `for` que usa a função `range` para exibir uma sequência de números.

```python
for i in range(3):
    print(i, end=' ')
```

Esta versão usa o argumento nomeado `end` para que a função `print` coloque um espaço depois de cada número, em vez de uma quebra de linha.

Também podemos usar um laço `for` para exibir as letras de uma string.

```python
for letter in 'Gadsby':
    print(letter, end=' ')
```

Observe que mudei o nome da variável de `i` para `letter`, o que dá mais informação sobre o valor a que ela se refere.
A variável definida em um laço `for` se chama **variável de laço**.

Agora que podemos percorrer as letras de uma palavra, podemos verificar se ela contém a letra "e".

```python
for letter in "Gadsby":
    if letter == 'E' or letter == 'e':
        print('This word has an "e"')
```

Antes de continuar, vamos encapsular esse laço em uma função.

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            print('This word has an "e"')
```

E vamos torná-la uma função pura que devolve `True` se a palavra contém um "e" e `False` caso contrário.

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            return True
    return False
```

Podemos generalizá-la para receber a palavra como parâmetro.

```python
def has_e(word):
    for letter in word:
        if letter == 'E' or letter == 'e':
            return True
    return False
```

Agora podemos testá-la assim:

```python
has_e('Gadsby')
```

```python
has_e('Emma')
```

## Atualizando variáveis

Como você talvez já tenha descoberto, é legal fazer mais de uma atribuição
à mesma variável.
Uma atribuição nova faz uma variável existente passar a se referir a um valor novo (e deixar de se referir ao valor antigo).

Por exemplo, aqui está uma atribuição inicial que cria uma variável.

```python
x = 5
x
```

E aqui está uma atribuição que muda o valor de uma variável.

```python
x = 7
x
```

A figura a seguir mostra como essas atribuições ficam em um diagrama de estado.

```python
from diagram import make_rebind, draw_bindings

bindings = make_rebind('x', [5, 7])
```

```python
from diagram import diagram, adjust

width, height, x, y = [0.54, 0.61, 0.07, 0.45]
ax = diagram(width, height)
bbox = draw_bindings(bindings, ax, x, y)
# adjust(x, y, bbox)
```

A seta pontilhada indica que `x` não se refere mais a `5`.
A seta sólida indica que agora ela se refere a `7`.

Um tipo comum de atribuição é uma **atualização**, em que o valor novo da
variável depende do antigo.

```python
x = 7
```

```python
x = x + 1
x
```

Esta instrução significa "pegue o valor atual de `x`, some um e atribua o resultado de volta a `x`."

Se você tentar atualizar uma variável que não existe, recebe um erro, porque o Python avalia a expressão da direita antes de atribuir um valor à variável da esquerda.

```python
%%expect NameError

z = z + 1
```

Antes de atualizar uma variável, você tem de **inicializá-la**, em geral
com uma atribuição simples:

```python
z = 0
z = z + 1
z
```

Aumentar o valor de uma variável se chama **incremento**; diminuir o valor se chama **decremento**.
Como essas operações são tão comuns, o Python oferece **operadores de atribuição aumentada** que atualizam uma variável de forma mais concisa.
Por exemplo, o operador `+=` incrementa uma variável pela quantidade dada.

```python
z += 2
z
```

Há operadores de atribuição aumentada para os outros operadores aritméticos, inclusive `-=` e `*=`.

## Percorrendo e contando

O programa a seguir conta o número de palavras na lista de palavras.

```python
total = 0

for line in open('words.txt'):
    word = line.strip()
    total += 1
```

Ele começa inicializando `total` com `0`.
A cada passagem pelo laço, incrementa `total` em `1`.
Assim, quando o laço termina, `total` se refere ao número total de palavras.

```python
total
```

Uma variável assim, usada para contar quantas vezes algo acontece, se chama **contador**.

Podemos acrescentar um segundo contador ao programa para acompanhar o número de palavras que contêm um "e".

```python
total = 0
count = 0

for line in open('words.txt'):
    word = line.strip()
    total = total + 1
    if has_e(word):
        count += 1
```

Vamos ver quantas palavras contêm um "e".

```python
count
```

Como porcentagem de `total`, cerca de dois terços das palavras usam a letra "e".

```python
count / total * 100
```

Assim você entende por que é difícil escrever um livro sem usar nenhuma dessas palavras.

## O operador in

A versão de `has_e` que escrevemos neste capítulo é mais complicada do que precisa ser.
O Python oferece um operador, `in`, que verifica se um caractere aparece em uma string.

```python
word = 'Gadsby'
'e' in word
```

Então podemos reescrever `has_e` assim.

```python
def has_e(word):
    if 'E' in word or 'e' in word:
        return True
    else:
        return False
```

E, como o condicional da instrução `if` tem um valor booleano, podemos eliminar a instrução `if` e devolver o booleano diretamente.

```python
def has_e(word):
    return 'E' in word or 'e' in word
```

Podemos simplificar esta função ainda mais usando o método `lower`, que converte as letras de uma string para minúsculas.
Aqui está um exemplo.

```python
word.lower()
```

`lower` cria uma string nova: ele não modifica a string existente, então o valor de `word` permanece inalterado.

```python
word
```

Aqui está como podemos usar `lower` em `has_e`.

```python
def has_e(word):
    return 'e' in word.lower()
```

```python
has_e('Gadsby')
```

```python
has_e('Emma')
```

## Busca

Com base nesta versão mais simples de `has_e`, vamos escrever uma função mais geral chamada `uses_any` que recebe um segundo parâmetro que é uma string de letras.
Ela devolve `True` se a palavra usa alguma das letras e `False` caso contrário.

```python
def uses_any(word, letters):
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

Aqui está um exemplo em que o resultado é `True`.

```python
uses_any('banana', 'aeiou')
```

E outro em que é `False`.

```python
uses_any('apple', 'xyz')
```

`uses_any` converte `word` e `letters` para minúsculas, então funciona com qualquer combinação de maiúsculas e minúsculas.

```python
uses_any('Banana', 'AEIOU')
```

A estrutura de `uses_any` é parecida com a de `has_e`.
Ela percorre as letras de `word` e as verifica uma de cada vez.
Se achar uma que aparece em `letters`, devolve `True` imediatamente.
Se chegar ao fim do laço sem achar nenhuma, devolve `False`.

Este padrão se chama **busca linear**.

## Doctest

No [Capítulo 4](section_docstring) usamos uma docstring para documentar uma função, isto é, para explicar o que ela faz.
Também é possível usar uma docstring para *testar* uma função.
Aqui está uma versão de `uses_any` com uma docstring que inclui testes.

```python
def uses_any(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any('banana', 'aeiou')
    True
    >>> uses_any('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

Cada teste começa com `>>>`, que é usado como prompt em alguns ambientes Python para indicar onde o usuário pode digitar código.
Em um doctest, o prompt é seguido de uma expressão, em geral uma chamada de função.
A linha seguinte indica o valor que a expressão deve ter se a função funcionar corretamente.

No primeiro exemplo, `'banana'` usa `'a'`, então o resultado deve ser `True`.
No segundo exemplo, `'apple'` não usa nenhuma de `'xyz'`, então o resultado deve ser `False`.

Para rodar esses testes, precisamos importar o módulo `doctest` e rodar uma função chamada `run_docstring_examples`.
Para tornar esta função mais fácil de usar, escrevi a função a seguir, que recebe um objeto de função como argumento.

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```

Ainda não aprendemos sobre `globals` e `__name__`: você pode ignorá-los.
Agora podemos testar `uses_any` assim.

```python
run_doctests(uses_any)
```

`run_doctests` acha as expressões na docstring e as avalia.
Se o resultado for o valor esperado, o teste **passa**.
Caso contrário, ele **falha**.

Se todos os testes passarem, `run_doctests` não exibe nada: nesse caso, nenhuma notícia é boa notícia.
Para ver o que acontece quando um teste falha, aqui está uma versão incorreta de `uses_any`.

```python
def uses_any_incorrect(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any_incorrect('banana', 'aeiou')
    True
    >>> uses_any_incorrect('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
        else:
            return False     # INCORRETO!
```

E aqui está o que acontece quando a testamos.

```python
run_doctests(uses_any_incorrect)
```

A saída inclui o exemplo que falhou, o valor que a função deveria produzir e o valor que a função de fato produziu.

Se você não tiver certeza de por que este teste falhou, terá a chance de depurá-lo como exercício.

## Glossário

**variável de laço:**
Uma variável definida no cabeçalho de um laço `for`.

**objeto de arquivo:**
Um objeto que representa um arquivo aberto e guarda quais partes do arquivo já foram lidas ou escritas.

**método:**
 Uma função associada a um objeto e chamada usando o operador ponto.

**atualização:**
Uma instrução de atribuição que dá um valor novo a uma variável que já existe, em vez de criar uma variável nova.

**inicializar:**
Criar uma variável nova e dar um valor a ela.

**incremento:**
Aumentar o valor de uma variável.

**decremento:**
Diminuir o valor de uma variável.

**contador:**
 Uma variável usada para contar alguma coisa, em geral inicializada com zero e depois incrementada.

**busca linear:**
Um padrão computacional que busca em uma sequência de elementos e para quando acha o que está procurando.

**passar:**
Se um teste roda e o resultado é o esperado, o teste passa.

**falhar:**
Se um teste roda e o resultado não é o esperado, o teste falha.
