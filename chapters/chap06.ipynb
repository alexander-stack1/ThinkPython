# Valores de retorno

Nos capítulos anteriores, usamos funções nativas, como `abs` e `round`, e funções do módulo math, como `sqrt` e `pow`.
Quando você chama uma dessas funções, ela devolve um valor que você pode atribuir a uma variável ou usar como parte de uma expressão.

As funções que escrevemos até agora são diferentes.
Algumas usam a função `print` para exibir valores, e algumas usam funções de tartaruga para desenhar figuras.
Mas elas não devolvem valores que atribuímos a variáveis ou usamos em expressões.

Neste capítulo, vamos ver como escrever funções que devolvem valores.

## Algumas funções têm valor de retorno

Quando você chama uma função como `math.sqrt`, o resultado se chama **valor de retorno**.
Se a chamada da função aparece no fim de uma célula, o Jupyter exibe o valor de retorno imediatamente.

```python
import math

math.sqrt(42 / math.pi)
```

Se você atribuir o valor de retorno a uma variável, ele não é exibido.

```python
radius = math.sqrt(42 / math.pi)
```

Mas você pode exibi-lo depois.

```python
radius
```

Ou pode usar o valor de retorno como parte de uma expressão.

```python
radius + math.sqrt(42 / math.pi)
```

Aqui está um exemplo de função que devolve um valor.

```python
def circle_area(radius):
    area = math.pi * radius**2
    return area
```

`circle_area` recebe `radius` como parâmetro e calcula a área de um círculo com esse raio.

A última linha é uma instrução `return` que devolve o valor de `area`.

Se chamarmos a função assim, o Jupyter exibe o valor de retorno.


```python
circle_area(radius)
```

Podemos atribuir o valor de retorno a uma variável.

```python
a = circle_area(radius)
```

Ou usá-lo como parte de uma expressão.

```python
circle_area(radius) + 2 * circle_area(radius / 2)
```

Depois podemos exibir o valor da variável à qual atribuímos o resultado.

```python
a
```

Mas não podemos acessar `area`.

```python
%%expect NameError

area
```

`area` é uma variável local de uma função, então não podemos acessá-la de fora da função.

## E algumas têm None

Se uma função não tem uma instrução `return`, ela devolve `None`, que é um valor especial como `True` e `False`.
Por exemplo, aqui está a função `repeat` do Capítulo 3.

```python
def repeat(word, n):
    print(word * n)
```

Se a chamarmos assim, ela exibe a primeira linha da canção "Finland", dos Monty Python.

```python
repeat('Finland, ', 3)
```

Esta função usa a função `print` para exibir uma string, mas não usa uma instrução `return` para devolver um valor.
Se atribuirmos o resultado a uma variável, ela exibe a string mesmo assim.

```python
result = repeat('Finland, ', 3)
```

E se exibirmos o valor da variável, não aparece nada.

```python
result
```

`result` de fato tem um valor, mas o Jupyter não o mostra.
No entanto, podemos exibi-lo assim.

```python
print(result)
```

O valor de retorno de `repeat` é `None`.

Agora aqui está uma função parecida com `repeat`, só que ela tem um valor de retorno.

```python
def repeat_string(word, n):
    return word * n
```

Observe que podemos usar uma expressão em uma instrução `return`, não só uma variável.

Com esta versão, podemos atribuir o resultado a uma variável.
Quando a função roda, ela não exibe nada.

```python
line = repeat_string('Spam, ', 4)
```

Mas depois podemos exibir o valor atribuído a `line`.

```python
line
```

Uma função como esta se chama **função pura** porque não exibe nada nem tem nenhum outro efeito, além de devolver um valor.

## Valores de retorno e condicionais

Se o Python não fornecesse `abs`, poderíamos escrevê-la assim.

```python
def absolute_value(x):
    if x < 0:
        return -x
    else:
        return x
```

Se `x` for negativo, a primeira instrução `return` devolve `-x` e a função termina imediatamente.
Caso contrário, a segunda instrução `return` devolve `x` e a função termina.
Então esta função está correta.

No entanto, se você colocar instruções `return` em um condicional, precisa garantir que todo caminho possível pelo programa encontre uma instrução `return`.
Por exemplo, aqui está uma versão incorreta de `absolute_value`.

```python
def absolute_value_wrong(x):
    if x < 0:
        return -x
    if x > 0:
        return x
```

Veja o que acontece se chamarmos esta função com `0` como argumento.

```python
absolute_value_wrong(0)
```

Não aparece nada! O problema é este: quando `x` é `0`, nenhuma das condições é verdadeira, e a função termina sem encontrar uma instrução `return`, o que significa que o valor de retorno é `None`, então o Jupyter não exibe nada.

Como outro exemplo, aqui está uma versão de `absolute_value` com uma instrução `return` extra no fim.

```python
def absolute_value_extra_return(x):
    if x < 0:
        return -x
    else:
        return x
    
    return 'This is dead code'
```

Se `x` for negativo, a primeira instrução `return` roda e a função termina.
Caso contrário, a segunda instrução `return` roda e a função termina.
De um jeito ou de outro, nunca chegamos à terceira instrução `return`: ela nunca pode rodar.

Código que nunca pode rodar se chama **código morto**.
Em geral, código morto não causa dano, mas costuma indicar um mal-entendido, e pode confundir quem tenta entender o programa.

## Desenvolvimento incremental

À medida que você escreve funções maiores, pode perceber que passa mais
tempo depurando.
Para lidar com programas cada vez mais complexos, vale tentar o **desenvolvimento incremental**, que é um jeito de acrescentar e testar só uma pequena quantidade de código de cada vez.

Como exemplo, suponha que você queira achar a distância entre dois pontos representados pelas coordenadas $(x_1, y_1)$ e $(x_2, y_2)$.
Pelo teorema de Pitágoras, a distância é:

$$\mathrm{distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$ 

O primeiro passo é pensar como uma função `distance` deveria parecer em Python, isto é, quais são as entradas (parâmetros) e qual é a saída (valor de retorno)?

Para esta função, as entradas são as coordenadas dos pontos.
O valor de retorno é a distância.
Na hora você já pode escrever um esboço da função:

```python
def distance(x1, y1, x2, y2):
    return 0.0
```

Esta versão ainda não calcula distâncias: ela sempre devolve zero.
Mas é uma função completa, com um valor de retorno, o que significa que você pode testá-la antes de torná-la mais complicada.

Para testar a função nova, vamos chamá-la com argumentos de exemplo:

```python
distance(1, 2, 4, 6)
```

Escolhi esses valores para que a distância horizontal seja `3` e a
distância vertical seja `4`.
Assim, o resultado é `5`, a hipotenusa de um triângulo retângulo `3-4-5`. Ao testar uma função, é útil saber a resposta certa.

Neste ponto já confirmamos que a função roda e devolve um valor, e podemos começar a acrescentar código no corpo.
Um bom próximo passo é achar as diferenças `x2 - x1` e `y2 - y1`. 
Aqui está uma versão que guarda esses valores em variáveis temporárias e os exibe.

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    print('dx is', dx)
    print('dy is', dy)
    return 0.0
```

Se a função estiver funcionando, ela deve exibir `dx is 3` e `dy is 4`.
Se isso acontecer, sabemos que a função está recebendo os argumentos certos e
executando o primeiro cálculo corretamente. Se não, há só algumas
linhas para conferir.

```python
distance(1, 2, 4, 6)
```

Até aqui, tudo bem. Em seguida calculamos a soma dos quadrados de `dx` e `dy`:

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    print('dsquared is: ', dsquared)
    return 0.0
```

De novo, podemos rodar a função e conferir a saída, que deve ser `25`.

```python
distance(1, 2, 4, 6)
```

Por fim, podemos usar `math.sqrt` para calcular a distância:

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    result = math.sqrt(dsquared)
    print("result is", result)
```

E testá-la.

```python
distance(1, 2, 4, 6)
```

O resultado está correto, mas esta versão da função exibe o resultado em vez de devolvê-lo, então o valor de retorno é `None`.

Podemos corrigir isso substituindo a função `print` por uma instrução `return`.

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    result = math.sqrt(dsquared)
    return result
```

Esta versão de `distance` é uma função pura.
Se a chamarmos assim, só o resultado é exibido.

```python
distance(1, 2, 4, 6)
```

E se atribuirmos o resultado a uma variável, nada é exibido.

```python
d = distance(1, 2, 4, 6)
```

As instruções `print` que escrevemos são úteis para depuração, mas depois que a função está funcionando, podemos removê-las. 
Código assim se chama **andaime** porque ajuda a construir o programa, mas não faz parte do produto final.

Este exemplo demonstra o desenvolvimento incremental.
Os pontos-chave deste processo são:

1.  Comece com um programa que funciona, faça mudanças pequenas e teste depois de cada mudança.

2.  Use variáveis para guardar valores intermediários, para poder exibi-los e conferi-los.

3.  Quando o programa estiver funcionando, remova o andaime.

Em qualquer momento, se houver um erro, você deve ter uma boa ideia de onde ele está.
O desenvolvimento incremental pode economizar muito tempo de depuração.

## Funções booleanas

Funções podem devolver os valores booleanos `True` e `False`, o que costuma ser conveniente para encapsular um teste complexo em uma função.
Por exemplo, `is_divisible` verifica se `x` é divisível por `y` sem resto.

```python
def is_divisible(x, y):
    if x % y == 0:
        return True
    else:
        return False
```

Aqui está como a usamos.

```python
is_divisible(6, 4)
```

```python
is_divisible(6, 3)
```

Dentro da função, o resultado do operador `==` é um booleano, então podemos escrever a
função de forma mais concisa devolvendo-o diretamente.

```python
def is_divisible(x, y):
    return x % y == 0
```

Funções booleanas costumam ser usadas em instruções condicionais.

```python
if is_divisible(6, 2):
    print('divisible')
```

Pode ser tentador escrever algo assim:

```python
if is_divisible(6, 2) == True:
    print('divisible')
```

Mas a comparação é desnecessária.

## Recursão com valores de retorno

Agora que podemos escrever funções com valores de retorno, podemos escrever funções recursivas com valores de retorno, e com essa capacidade cruzamos um limiar importante: o subconjunto de Python que temos agora é **Turing-completo**, o que significa que podemos realizar qualquer computação que possa ser descrita por um algoritmo.

Para demonstrar a recursão com valores de retorno, vamos avaliar algumas funções matemáticas definidas de forma recursiva.
Uma definição recursiva é parecida com uma definição circular, no sentido de que a definição se refere à coisa que está sendo definida. Uma definição realmente circular não é muito útil:

> vorpal: Adjetivo usado para descrever algo que é vorpal.

Se você visse essa definição no dicionário, provavelmente ficaria irritado. 
Por outro lado, se procurasse a definição da função fatorial, denotada pelo símbolo $!$, poderia encontrar algo assim: 

$$\begin{aligned}
0! &= 1 \\
n! &= n~(n-1)!
\end{aligned}$$ 

Esta definição diz que o fatorial de $0$ é $1$, e o fatorial de qualquer outro valor, $n$, é $n$ multiplicado pelo fatorial de $n-1$.

Se você consegue escrever uma definição recursiva de alguma coisa, consegue escrever um programa em Python para avaliá-la. 
Seguindo um processo de desenvolvimento incremental, vamos começar com uma função que recebe `n` como parâmetro e sempre devolve `0`.

```python
def factorial(n):
    return 0
```

Agora vamos acrescentar a primeira parte da definição: se o argumento for `0`, só precisamos devolver `1`:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return 0
```

Agora vamos preencher a segunda parte: se `n` não for `0`, precisamos fazer uma chamada
recursiva para achar o fatorial de `n-1` e depois multiplicar o resultado por `n`:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        recurse = factorial(n-1)
        return n * recurse
```

O fluxo de execução deste programa é parecido com o fluxo de `countdown` no Capítulo 5.
Se chamarmos `factorial` com o valor `3`:

Como `3` não é `0`, tomamos o segundo ramo e calculamos o fatorial
de `n-1`\...

> Como `2` não é `0`, tomamos o segundo ramo e calculamos o
> fatorial de `n-1`\...
>
> > Como `1` não é `0`, tomamos o segundo ramo e calculamos o
> > fatorial de `n-1`\...
> >
> > > Como `0` é igual a `0`, tomamos o primeiro ramo e devolvemos `1` sem
> > > fazer mais nenhuma chamada recursiva.
> >
> > O valor de retorno, `1`, é multiplicado por `n`, que é `1`, e o
> > resultado é devolvido.
>
> O valor de retorno, `1`, é multiplicado por `n`, que é `2`, e o resultado
> é devolvido.

O valor de retorno `2` é multiplicado por `n`, que é `3`, e o resultado,
`6`, torna-se o valor de retorno da chamada de função que iniciou o
processo inteiro.

A figura a seguir mostra o diagrama de pilha desta sequência de chamadas de função.

```python
from diagram import Frame, Stack, make_binding

main = Frame([], name='__main__', loc='left')
frames = [main]

ns = 3, 2, 1
recurses = 2, 1, 1
results = 6, 2, 1

for n, recurse, result in zip(ns, recurses, results):
    binding1 = make_binding('n', n)
    binding2 = make_binding('recurse', recurse)
    frame = Frame([binding1, binding2], 
                  name='factorial', value=result,
                  loc='left', dx=1.2)
    frames.append(frame)
    
binding1 = make_binding('n', 0)
frame = Frame([binding1], name='factorial', value=1, 
              shim=1.2, loc='left', dx=1.4)
frames.append(frame)

stack = Stack(frames, dy=-0.45)
```

```python
from diagram import diagram, adjust

width, height, x, y = [2.74, 2.26, 0.73, 2.05]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)
```

Os valores de retorno aparecem sendo passados de volta pela pilha.
Em cada quadro, o valor de retorno é o produto de `n` e `recurse`.

No último quadro, a variável local `recurse` não existe porque o ramo que a cria não roda.

## Salto de fé

Seguir o fluxo de execução é um jeito de ler programas, mas isso pode se tornar esmagador depressa. Uma alternativa é o que chamo de "salto de fé". Quando você chega a uma chamada de função, em vez de seguir o fluxo de execução, você *assume* que a função funciona corretamente e devolve o resultado certo.

Na verdade, você já pratica esse salto de fé quando usa funções nativas.
Quando chama `abs` ou `math.sqrt`, você não examina o corpo dessas funções: só assume que elas funcionam.

O mesmo vale quando você chama uma de suas próprias funções. Por exemplo, antes escrevemos uma função chamada `is_divisible` que determina se um número é divisível por outro. Depois que nos convencemos de que esta função está correta, podemos usá-la sem olhar o corpo de novo.

O mesmo vale para programas recursivos.
Quando você chega à chamada recursiva, em vez de seguir o fluxo de execução, deve assumir que a chamada recursiva funciona e então se perguntar: "Assumindo que consigo calcular o fatorial de $n-1$, consigo calcular o fatorial de $n$?"
A definição recursiva do fatorial implica que sim, multiplicando por $n$.

Claro, é um pouco estranho assumir que a função funciona corretamente quando você ainda não terminou de escrevê-la, mas é por isso que se chama salto de fé!

## Fibonacci

Depois de `factorial`, o exemplo mais comum de função recursiva é `fibonacci`, que tem a seguinte definição: 

$$\begin{aligned}
\mathrm{fibonacci}(0) &= 0 \\
\mathrm{fibonacci}(1) &= 1 \\
\mathrm{fibonacci}(n) &= \mathrm{fibonacci}(n-1) + \mathrm{fibonacci}(n-2)
\end{aligned}$$ 

Traduzido para Python, fica assim:

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif  n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

Se você tentar seguir o fluxo de execução aqui, mesmo para valores pequenos de $n$, a cabeça explode.
Mas, segundo o salto de fé, se você assume que as duas chamadas recursivas funcionam corretamente, pode ter confiança de que a última instrução `return` está correta.

Como observação à parte, este jeito de calcular números de Fibonacci é muito ineficiente.
No [Capítulo 10](section_memos) explico por quê e sugiro um jeito de melhorar.

## Verificando tipos

O que acontece se chamarmos `factorial` e passarmos `1.5` como argumento?

```python
%%expect RecursionError

factorial(1.5)
```

Parece uma recursão infinita. Como isso é possível? A função tem casos-base quando `n == 1` ou `n == 0`.
Mas se `n` não for um inteiro, podemos *perder* o caso-base e recursar para sempre.

Neste exemplo, o valor inicial de `n` é `1.5`.
Na primeira chamada recursiva, o valor de `n` é `0.5`.
Na seguinte, é `-0.5`. 
A partir daí, ele fica menor (mais negativo), mas nunca será `0`.

Para evitar a recursão infinita, podemos usar a função nativa `isinstance` para verificar o tipo do argumento.
Aqui está como verificamos se um valor é um inteiro.

```python
isinstance(3, int)
```

```python
isinstance(1.5, int)
```

Agora aqui está uma versão de `factorial` com verificação de erros.

```python
def factorial(n):
    if not isinstance(n, int):
        print('factorial is only defined for integers.')
        return None
    elif n < 0:
        print('factorial is not defined for negative numbers.')
        return None
    elif n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

Primeiro ela verifica se `n` é um inteiro.
Se não for, exibe uma mensagem de erro e devolve `None`.



```python
factorial('crunchy frog')
```

Depois verifica se `n` é negativo.
Se for, exibe uma mensagem de erro e devolve `None.`

```python
factorial(-2)
```

Se passarmos pelas duas verificações, sabemos que `n` é um inteiro não negativo, então podemos ter confiança de que a recursão vai terminar.
Verificar os parâmetros de uma função para garantir que tenham os tipos e valores corretos se chama **validação de entrada**.

## Depuração

Quebrar um programa grande em funções menores cria pontos de verificação naturais para a depuração.
Se uma função não estiver funcionando, há três possibilidades a considerar:

-   Há algo errado com os argumentos que a função está recebendo, isto é, uma pré-condição foi violada.

-   Há algo errado com a função, isto é, uma pós-condição foi violada.

-   Quem chamou está fazendo algo errado com o valor de retorno.

Para descartar a primeira possibilidade, você pode acrescentar uma instrução `print` no início da função que exiba os valores dos parâmetros (e talvez os tipos deles).
Ou pode escrever código que verifique as pré-condições de forma explícita.

Se os parâmetros parecerem bons, você pode acrescentar uma instrução `print` antes de cada instrução `return` e exibir o valor de retorno.
Se possível, chame a função com argumentos que tornem fácil conferir o resultado. 

Se a função parecer estar funcionando, olhe a chamada da função para garantir que o valor de retorno esteja sendo usado corretamente, ou usado de algum jeito!

Acrescentar instruções `print` no início e no fim de uma função pode ajudar a tornar o fluxo de execução mais visível.
Por exemplo, aqui está uma versão de `factorial` com instruções print:

```python
def factorial(n):
    space = ' ' * (4 * n)
    print(space, 'factorial', n)
    if n == 0:
        print(space, 'returning 1')
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        print(space, 'returning', result)
        return result
```

`space` é uma string de caracteres de espaço que controla a indentação da
saída. Aqui está o resultado de `factorial(3)` :

```python
factorial(3)
```

Se você estiver confuso sobre o fluxo de execução, este tipo de saída pode ajudar.
Leva algum tempo desenvolver um andaime eficaz, mas um pouco de andaime pode economizar muita depuração.

## Glossário

**valor de retorno:**
O resultado de uma função. Se uma chamada de função é usada como expressão, o valor de retorno é o valor da expressão.

**função pura:**
Uma função que não exibe nada nem tem nenhum outro efeito, além de devolver um valor de retorno.


**código morto:**
Parte de um programa que nunca pode rodar, muitas vezes porque aparece depois de uma instrução `return`.

**desenvolvimento incremental:**
Um plano de desenvolvimento de programa destinado a evitar a depuração, acrescentando e testando só uma pequena quantidade de código de cada vez.

**andaime:**
 Código usado durante o desenvolvimento do programa, mas que não faz parte da versão final.

**Turing-completo:**
Uma linguagem, ou um subconjunto de uma linguagem, é Turing-completo se puder realizar qualquer computação que possa ser descrita por um algoritmo.

**validação de entrada:**
Verificar os parâmetros de uma função para garantir que tenham os tipos e valores corretos
